### BeautifulSoup
: 간단하게 HTML과 XML에서 정보를 추출

### BeautifulSoup 기본 사용법

```python
from bs4 import BeautifulSoup
import urllib.request as req

url = "https://zeushahn.github.io/Test/python/bs_exam01.html"

res = req.urlopen(url)
res.status


# BeautifulSoup으로 분석하기
soup = BeautifulSoup(res, "html.parser")
print(soup)
```

```python
# 원하는 부분 추출하기
h1 = soup.html.body.h1 ##Tag이름으로 가져온다
p1 = soup.html.body.p
p2 = p1.next_sibling.next_sibling ## </p>도 인식하기 때문에 두번 적어야 한다

# 요소의 글자 출력하기
print("h1 =", h1.string, h1.text, h1.get_text())
print("p =", p1.string)
print("p =", p2.string)
```

### id로 요소 찾는 방법

```python
from bs4 import BeautifulSoup
import urllib.request as req

url = "https://zeushahn.github.io/Test/python/bs_exam02.html"

res = req.urlopen(url)
res.status


# BeautifulSoup으로 분석하기
soup = BeautifulSoup(res, "html.parser")
print(soup)
```

```python
# find() Method 로 원하는 부분 추출하기

title = soup.find(id='title')
body = soup.find(id='body')

# 출력하기
print("title =", title.string)
print("body =",body.string)
```

### 여러개의 요소 추출하기

```python
from bs4 import BeautifulSoup
import urllib.request as req

url = "https://zeushahn.github.io/Test/python/bs_exam03.html"

res = req.urlopen(url)
res.status


# BeautifulSoup으로 분석하기
soup = BeautifulSoup(res, "html.parser")
print(soup)
```

### CSS 선택자 사용하기
: BeautifulSoup는 자바 스크립트 라이브러리인 JQuery처럼 CSS 선택자를 지정해서 요소를 추출 가능

```python
from bs4 import BeautifulSoup
import urllib.request as req

url = "https://zeushahn.github.io/Test/python/bs_exam04.html"

res = req.urlopen(url)
res.status


# BeautifulSoup으로 분석하기
soup = BeautifulSoup(res, "html.parser")
print(soup)
```

```python
# 필요한 부분을 CSS쿼리로 추출하기 (#:id, .:class, > : 자식, 빈칸 : 후손)
# 타이틀 하나만 부분 추출하기
h1 = soup.select_one("div#meigen > h1").string
print("h1 :", h1)

# 목록 부분 여러개 추출하기
li_list = soup.select("div#meigen > ul.items > li")

for li in li_list:
  print("li =", li.string)
```

```python
from bs4 import BeautifulSoup
import urllib.request as req
import urllib.parse
import time

title = []
contents = []

url = "https://news.daum.net/tech"

res = req.urlopen(url)
# res.status

# BeautifulSoup으로 분석하기
soup = BeautifulSoup(res, "html.parser")
a_list = soup.select(".item_newsheadline2")
a_list = a_list[:9]
# print(a_list)

for a in a_list:
    print(a.attrs['href'], ":", end=" ")
    print(urllib.parse.unquote(a.attrs['data-title']))
    title.append(urllib.parse.unquote(a.attrs['data-title']))
    sub_url = a.attrs['href']
    sub_res = req.urlopen(sub_url)
    sub_soup = BeautifulSoup(sub_res, "html.parser")
    sub_a_list = sub_soup.select("div.article_view > section > p")
    
    # print(sub_a_list)
    #mArticle > div.news_view.fs_type1 > div.article_view > section > p:nth-child(3)
    temp_contents = ""
    for sub_a in sub_a_list:
        temp_contents += sub_a.text
    contents.append(temp_contents)
    time.sleep(1)
```
