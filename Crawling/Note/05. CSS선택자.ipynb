#### 위키문헌에 공개되어 있는 윤동주 시인의 작품 목록 가져오기

```python
# parse를 사용하여 인코딩
from urllib import parse
from bs4 import BeautifulSoup
import urllib.request as req
import time

auth = "저자"
name = "윤동주"
url = "https://ko.wikisource.org/wiki/"
url = url + parse.quote(auth) + ":" + parse.quote(name)

# print(url)

res = req.urlopen(url)
# res.status

soup = BeautifulSoup(res, "html.parser")
a_list = soup.select(".mw-parser-output li a")
# print(a_list)
a_list = a_list[3:]


for a in a_list:
    print(a.string)
    


```
