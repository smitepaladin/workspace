```python
from bs4 import BeautifulSoup
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from webdriver_manager.chrome import ChromeDriverManager
import time
import pandas as pd
```

```python
# 드라이버 설정
driver = webdriver.Chrome(service=Service(ChromeDriverManager().install()))
```

```python
companies = []

# 2. 페이지 순회
for page in range(1, 6):
    url = f"https://www.jobkorea.co.kr/Salary/Index?coPage={page}"
    driver.get(url)

    try:
        WebDriverWait(driver, 10).until(
            EC.presence_of_element_located((By.CSS_SELECTOR, "#listCompany > li"))
        )
        time.sleep(1)

        html = driver.page_source
        soup = BeautifulSoup(html, "html.parser")

        items = soup.select("#listCompany > li")

        for item in items:
            name = item.select_one("div.headers > div")
            salary = item.select_one("div.salary strong")
            count = item.select_one("div.summary > div:nth-child(2)")

            companies.append({
                "회사명": name.text.strip() if name else "",
                "연봉": salary.text.strip() if salary else "",
                "사원수": count.text.strip() if count else ""
            })

        print(f"{page}페이지 완료")

    except Exception as e:
        print(f"{page}페이지 에러: {e}")

driver.quit()
```

```python


# 데이터프레임으로 변환 및 확인
df = pd.DataFrame(companies)
print(df)

# 필요시 CSV 저장
# df.to_csv("jobkorea_salary_data.csv", index=False, encoding="utf-8-sig")
```
