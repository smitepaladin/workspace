```python
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
import pandas as pd
import time

# Chrome 옵션 설정
options = Options()
options.add_argument("--headless=new")  # 중요: 안정적인 headless
options.add_argument("--disable-gpu")
options.add_argument("--no-sandbox")
options.add_argument("--disable-dev-shm-usage")
options.add_argument("--window-size=1920,1080")

# 드라이버 실행
driver = webdriver.Chrome(options=options)

try:
    url = "https://www.work24.go.kr/wk/a/b/1200/retriveDtlEmpSrchList.do?...&currentPageNo=1&pageIndex=1"
    driver.get(url)
    time.sleep(3)

    company_list = []
    salary_list = []
    career_list = []
    edu_list = []

    companies = driver.find_elements(By.CSS_SELECTOR, "#list1 > td.al_left.pd24 > div > div:nth-child(1) > div > label > span > a")
    salaries = driver.find_elements(By.CSS_SELECTOR, "#list1 > td.link.pd24 > div > ul > li.dollar > p > span")
    careers = driver.find_elements(By.CSS_SELECTOR, "#list1 > td.link.pd24 > div > ul > li.member > p > span:nth-child(1)")
    edus = driver.find_elements(By.CSS_SELECTOR, "#list1 > td.link.pd24 > div > ul > li.member > p > span:nth-child(2)")

    count = min(len(companies), len(salaries), len(careers), len(edus))

    for i in range(count):
        try:
            company_list.append(companies[i].text.strip())
            salary_list.append(salaries[i].text.strip())
            career_list.append(careers[i].text.strip())
            edu_list.append(edus[i].text.strip())
            print(f"{i} {companies[i].text.strip()} {salaries[i].text.strip()} {careers[i].text.strip()} {edus[i].text.strip()}")
        except Exception as e:
            print(f"[오류] {i}번째 항목에서 에러 발생:", e)

    # 저장
    df = pd.DataFrame({
        "회사명": company_list,
        "연봉": salary_list,
        "경력": career_list,
        "학력": edu_list
    })
    df.to_csv("work24_ai_simple.csv", index=False, encoding="utf-8-sig")

except Exception as e:
    print("❌ 전체 오류 발생:", e)

finally:
    driver.quit()
```
