```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager
from selenium.webdriver.common.by import By
```

```python
# Chrome Browser와 Chrome Driver Version 확인
chrome_options = webdriver.ChromeOptions()
driver = webdriver.Chrome(
  service=Service(ChromeDriverManager().install()),
  options=chrome_options
)
driver.get("https://www.opinet.co.kr/searRgSelect.do")
```

```python
# 현재 화면 Capture하기
driver.save_screenshot("../Data/001.png")
```

### 서울시 유가 정보 획득을 위한 Test

```python
# 자치구명 정보 가져오기 : //*[@id="SIGUNGU_NM0"]

gu_list_raw = driver.find_element(By.XPATH, '//*[@id="SIGUNGU_NM0"]')
gu_list = gu_list_raw.find_elements(By.TAG_NAME, 'option')

gu_name = [option.get_attribute('value') for option in gu_list]
gu_name.remove("")
gu_name
```

```python
# 조회버튼 click ; //*[@id="searRgSelect"]/span
xpath = '//*[@id="searRgSelect"]/span'
element_sel_gu = driver.find_element(By.XPATH, xpath).click()
```

```python
# 자치구 이름을 획득하여 1번 자치구를 실행해보자
element = driver.find_element(By.ID, "SIGUNGU_NM0")
element.send_keys(gu_name[0])
```

```python
# XPATH를 사용하여 조회버튼 Click
xpath = '//*[@id="searRgSelect"]/span'
element_sel_gu = driver.find_element(By.XPATH, xpath).click()
```

```python
### xpath를 사용하여 엑셀 저장 버튼 click : //*[@id="templ_list0"]/div[7]/div/a/span
xpath = '//*[@id="templ_list0"]/div[7]/div/a/span'
element_get_excel = driver.find_element(By.XPATH, xpath).click()
```

### 데이터 수집

```python
import time
for gu in gu_name:
  element = driver.find_element(By.ID, "SIGUNGU_NM0")
  element.send_keys(gu)
  time.sleep(7)

  xpath = '//*[@id="searRgSelect"]/span'
  element_sel_gu = driver.find_element(By.XPATH, xpath).click()
  time.sleep(10)

  xpath = '//*[@id="templ_list0"]/div[7]/div/a/span'
  element_get_excel = driver.find_element(By.XPATH, xpath).click()
  time.sleep(7)
```
