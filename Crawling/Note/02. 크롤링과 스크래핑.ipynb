### 크롤링과 스크래핑

#### requests
##### 응답코드
1XX : 요청을 받았고 작업 진행중.


2XX : 사용자의 요청이 성공적으로 수행됨.


3XX : 요청은 완료되었으나 리다이렉션이 필요. (우리 말고 다른쪽 서버도 필요)


4XX : 사용자의 요청이 잘못됨. (대부분 파일이 없음)


5XX : 서버에 오류가 발생함.

```python
# request 응답 코드

import requests

URL = "https://www.naver.com"
response = requests.get(URL)
print(response.status_code)
print(response.text)
```

```python
# 네이버에서 정보 검색 결과 가져오기
# https://comic.naver.com/webtoon/list?titleId=747269

import requests

URL = "https://comic.naver.com/webtoon/list"
params = {'titleId' : '747269'}
response = requests.get(URL, params=params)
print(response.status_code)
print(response.text)
```

```python
# Json형식의 데이터 가져오기
import requests
response = requests.get("https://raw.githubusercontent.com/naver/naver-openapi-guide/draft/naver-openapi-swagger.json")
result = response.json()
print(result)
print(type(result))
print(result['info']['description'])
```
