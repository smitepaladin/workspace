### 웹상의 정보를 추출하는 방법
- urllib 라이브러리 사용하여 HTTP 또는 FTP를 사용해 데이터를 다운로드


```python
# 라이브러리
import urllib.request as request

# 다운로드 파일이름 결정
fileName = "mov01.jpg"

# URL과 저장 경로 지정하기
url = "https://zeushahn.github.io/Test/images/" + fileName
saveName = "../Data/" + fileName

# 다운로드
request.urlretrieve(url, saveName)
print("OK")
```

```python
# Ex : mov02.jpg, mov03.jpg, mov04.jpg

import urllib.request as request

for i in range(1, 4+1):  # 1부터 4까지
    fileName = f"mov0{str(i)}.jpg" 
    url = "https://zeushahn.github.io/Test/images/" + fileName
    saveName = "../Data/" + fileName

    request.urlretrieve(url, saveName)
    print(f"{fileName} 다운로드 완료")
```

```python
# Ex : mov02.jpg, mov03.jpg, mov04.jpg

fileLists = ['mov02.jpg', 'mov03.jpg', 'mov04.jpg']

for fileName in fileLists:
  url = "https://zeushahn.github.io/Test/images/" + fileName
  saveName = "../Data/" + fileName
  request.urlretrieve(url, saveName)
  print(fileName, "is downloaded.")
```

```python
# Ex : mov02.jpg, mov03.jpg, mov04.jpg

list(range(len(fileLists)))

fileLists = ['mov02.jpg', 'mov03.jpg', 'mov04.jpg']

for i in range(len(fileLists)):
  fileName = fileLists[i]
  url = "https://zeushahn.github.io/Test/images/" + fileName
  saveName = "../Data/" + fileName
  request.urlretrieve(url, saveName)
  print(fileName, "is downloaded.")
```
