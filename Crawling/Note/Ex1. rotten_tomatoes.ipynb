```python
# Module
from bs4 import BeautifulSoup
from selenium import webdriver
import urllib.request as req
import os
import time
```

```python
driver = webdriver.Chrome()
```

```python
driver.get("https://www.rottentomatoes.com/browse/movies_in_theaters/sort:popular?page=4")
time.sleep(3)
```

```python
# 크롤링한 데이터를 담을 리스트 생성
detail_links = []
titles = []
running_times = []
directors = []
in_theaters = []
distributors = []
```

```python
html = driver.page_source ## 셀레늄의 드라이버 아래 페이지 소스가 가지고 있다.
soup = BeautifulSoup(html, "html.parser")
a_list = soup.select("tile-dynamic > a")

# a_list = soup.select("tile-dynamic > rt-img")
# a_list = a_list[10:]
# # a_list = a_list[1:10]


# 1. 정보를 가져올 리뷰페이지 주소 수집
for a in a_list:
    detail_links.append("https://www.rottentomatoes.com" + a.attrs['href'] + "/reviews")
# for link in detail_links:
#     print(link)

# 디테일 페이지 순회
for url in detail_links[:99]:  # ← 너무 많으면 차단될 수 있으니 먼저 10개로 테스트
    driver.get(url)
    time.sleep(5)  # 로딩 대기

    detail_soup = BeautifulSoup(driver.page_source, "html.parser")
    # 2. 제목 수집
    
    title = detail_soup.select_one("#main-page-content > div > aside > section > div.info-section > a.sidebar-title")
    titles.append(title.text.strip())
# for title in titles:
#     print(title)

    # 3. 러닝타임 수집
    
    running_time = detail_soup.select_one("#main-page-content > div > aside > section > div.info-section > ul > li:nth-child(1)")
    running_times.append(running_time.text.strip()[6:])

    # 4. 감독 수집
    
    director = detail_soup.select_one("#main-page-content > div > aside > section > div.info-section > ul > li:nth-child(3) > a")
    directors.append(director.text.strip())

    # 5. 개봉일 수집
    
    in_theater = detail_soup.select_one("#main-page-content > div > aside > section > div.info-section > ul > li:nth-child(4)")
    in_theaters.append(in_theater.text.strip()[33:])

    # 6. 배급사 수집
    
    distributor = detail_soup.select_one("#main-page-content > div > aside > section > div.info-section > ul > li:nth-child(5)")
    distributors.append(distributor.text.strip())


    print(f"{titles[-1]} 완료")

driver.quit()
```

```python
# 결과물 찍어보기
for i in range(len(titles)):
    print(f"{i+1}. {titles[i]}")
    print(f"리뷰링크: {detail_links[i]}")
    print(f"러닝타임: {running_times[i]}")
    print(f"감독: {directors[i]}")
    print(f"개봉일: {in_theaters[i]}")
    print(f"배급사: {distributors[i]}")
    print("")
```
