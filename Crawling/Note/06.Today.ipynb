```python
from bs4 import BeautifulSoup
import urllib.request as req
import urllib.parse as parse
import datetime

now = datetime.datetime.now()
nowDate = now.strftime("%Y년 %m월 %d일 %H시 %M분 입니다.")


areas = ["서울", "대구", "부산", "제주도"]
whether = "날씨"

print("오늘의 요점 사항 입니다.")
print(f"현재 일자 및 시간은 {nowDate}")
print("**** 날씨정보 ****")

for area in areas:
    url = "https://search.naver.com/search.naver?where=nexearch&sm=top_hty&fbm=0&ie=utf8&query="
    url = url + parse.quote(area) + "+" + parse.quote(whether)

    res = req.urlopen(url)
    soup = BeautifulSoup(res, "html.parser")

    a_list = soup.select("div._today > div.weather_graphic > div.temperature_text > strong")
    b_list = soup.select("div._today > div.temperature_info > p > span.temperature.up")
    c_list = soup.select("div._today > div.temperature_info > p > span.weather.before_slash")

    a_text = a_list[0].text[5:]
    b_text = b_list[0].text
    c_text = c_list[0].text

    print(f"현재 {area}의 날씨 : {a_text}이고 어제보다 {b_text} {c_text}")


print("**** 주요 IT News ****")

url2 = "https://news.daum.net/"
res2 = req.urlopen(url2)
soup2 = BeautifulSoup(res2, "html.parser")

d_list = soup2.select(".item_newsheadline2")
d_list = d_list[:9]

for d in d_list:
    print(d.attrs['href'], ":", end=" ")
    print(parse.unquote(d.attrs['data-title']))
```
