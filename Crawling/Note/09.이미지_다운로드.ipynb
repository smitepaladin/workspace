### 이미지 다운로드용 웹크롤러 만들기
: https://pixabay.com/ko/

```python
# Module
from bs4 import BeautifulSoup
from selenium import webdriver
from selenium.webdriver.common.by import By
import urllib
import os
import math
import time
```

```python
query_text = input("1. 크롤링할 이미지의 키워드는 무엇 입니까? :") ## 기본적으로 str
cnt = int(input("2. 크롤링 할 건수는 몇건 입니까? :"))
real_cnt = math.ceil(cnt / 100) # 실제 크롤링 할 페이지수
f_dir = input("3. 파일이 저장될 경로만 쓰세요(예: ./Pixavbay :")
if f_dir == '':
  f_dir = "./Pixabay"

  print("")
  print("요청하신 데이터를 수집중이오니 잠시만 기다려주세요!!")
```

```python
# 파일을 저장할 폴더 생성
img_dir = "../Data/" + f_dir+"_"+query_text+"/"
os.makedirs(img_dir)
os.chdir(img_dir)
```

```python
# 크롬 드라이버를 사용해서 웹 브라우저 실행하기
driver = webdriver.Chrome()
driver.get("https://pixabay.com/ko/")
time.sleep(3)
```

```python
# 검색어 입력창에 검색어 입력후 검색 수행

element = driver.find_element(By.XPATH, '//*[@id="app"]/div[1]/div[1]/div[2]/div[2]/div[2]/form/div[1]/input')
element.send_keys(query_text)
element.submit() ## 엔터
```

```python
# 저장소 만들기
file_no = 1
count = 1
img_src = [] # 이미지 파일의 url 주소 저장용 List
```

```python
# 스크롤 다운 함수 만들기
def scroll_down(driver):
  driver.execute_script("window.scrollBy(0, 1000)")
  time.sleep(1)
```

```python
# 저장소 만들기
file_no = 1
count = 1
img_src = [] # 이미지 파일의 url 주소 저장용 List

for a in range(1, real_cnt+1):
    for b in range(0, 5):
        scroll_down(driver)
        time.sleep(1)
    
    # 원본 이미지 url 주소 수집
    html = driver.page_source
    soup = BeautifulSoup(html, 'html.parser')
    img_src_t = soup.find('div', 'results--mB75j').find_all('img')
    # print(img_src)
    
    for c in img_src_t:
        img_src1 = c['src']
        if 'http' in img_src1:
            img_src.append(img_src1)
            print(img_src1)
        count += 1
        if count > cnt:
            break
    
    # 수집된 url주소로 이미지 파일 가져와서 저장하기
    for e in range(0, len(img_src)):
        class AppURLopener(urllib.request.FancyURLopener):
            version = "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36"

        urllib.urlopener = AppURLopener()
        urllib.urlopener.retrieve(img_src[e], str(file_no)+".jpg")
        print(f"==>{file_no}")
        time.sleep(1)

        if file_no >= cnt:
            break

        file_no+=1

        if a > real_cnt:
            break
```
