```python
# Module
from bs4 import BeautifulSoup
from selenium import webdriver
import urllib.request as req
import os
import time
```

```python
# 파일을 저장할 폴더를 생성
os.mkdir("../Data/더조은IT아카데미/")
os.chdir("../Data/더조은IT아카데미")
```

```python
driver = webdriver.Chrome()
```

```python
driver.get("https://tjoeun.co.kr/front/lecture/lecture_list.asp?pg=1&ci=3&job=0")
time.sleep(2)
```

```python
# Beautiful Soup로 본문 내용만 출력하기
course = []
html = driver.page_source ## 셀레늄의 드라이버 아래 페이지 소스가 가지고 있다.
soup = BeautifulSoup(html, "html.parser")
a_list = soup.select("h3")

for a in a_list:
  print(a.text.strip())
```
