```python
def scroll_down(driver, pause_time=1.5):
    prev_height = driver.execute_script("return document.body.scrollHeight")
    while True:
        driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
        time.sleep(pause_time)
        curr_height = driver.execute_script("return document.body.scrollHeight")
        if curr_height == prev_height:
            break
        prev_height = curr_height
```

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from bs4 import BeautifulSoup
import pandas as pd
import time

driver = webdriver.Chrome()

driver.get("https://www.work24.go.kr/wk/a/b/1200/retriveDtlEmpSrchList.do?basicSetupYn=&careerTo=&keywordJobCd=&occupation=02&seqNo=&cloDateEndtParam=&payGbn=Y&templateInfo=&rot2WorkYn=&shsyWorkSecd=&resultCnt=10&keywordJobCont=Y&cert=&moreButtonYn=Y&minPay=1&codeDepth2Info=11000&currentPageNo=1&eventNo=&mode=&major=&resrDutyExcYn=&eodwYn=&sortField=DATE&staArea=&sortOrderBy=DESC&keyword=Ai&termSearchGbn=&carrEssYns=&benefitSrchAndOr=O&occupationParam=02&disableEmpHopeGbn=&actServExcYn=&keywordStaAreaNm=&maxPay=&emailApplyYn=&codeDepth1Info=11000&keywordEtcYn=&regDateStdtParam=&publDutyExcYn=&keywordJobCdSeqNo=&viewType=&exJobsCd=&templateDepthNmInfo=&region=&employGbn=&empTpGbcd=1&computerPreferential=&infaYn=&cloDateStdtParam=&siteClcd=all&searchMode=Y&birthFromYY=&indArea=&careerTypes=&subEmpHopeYn=&tlmgYn=&academicGbn=&templateDepthNoInfo=&foriegn=&entryRoute=&mealOfferClcd=&basicSetupYnChk=&station=&holidayGbn=&srcKeyword=Ai&academicGbnoEdu=noEdu&enterPriseGbn=&cloTermSearchGbn=&birthToYY=&keywordWantedTitle=&stationNm=&benefitGbn=&keywordFlag=&notSrcKeyword=&essCertChk=&depth2SelCode=&keywordBusiNm=&preferentialGbn=&rot3WorkYn=&regDateEndtParam=&pfMatterPreferential=&pageIndex=1&termContractMmcnt=&careerFrom=&laborHrShortYn=#scrollLoc")  # 그대로 사용하세요

# 각 열 데이터를 담을 리스트 초기화
results = []

# 1~6페이지 순회
for i in range(1, 7):  # 버튼은 1~6번 존재
    xpath = f'//*[@id="mForm"]/div[2]/div/div[2]/div/div/div/button[{i}]'
    element = driver.find_element(By.XPATH, xpath)

    driver.execute_script("arguments[0].scrollIntoView(true);", element)
    time.sleep(0.5)

    driver.execute_script("arguments[0].click();", element)
    time.sleep(1.5)

    html = driver.page_source
    soup = BeautifulSoup(html, "html.parser")

    # 데이터 추출
    names = soup.select("a.cp_name")
    pays = soup.select("li.dollar span")
    personal_historys = soup.select("li.member > p > span:nth-child(1)")
    titles = soup.select("td.al_left.pd24 div:nth-child(2) > a")
    registration_dates = soup.select("td:nth-child(3) > p:nth-child(4)")
    deadlines = soup.select("td:nth-child(3) > p:nth-child(3)")

    # 추출 결과 정제 및 리스트에 저장
    for company, salary, history, titlee, r, d in zip(
        names, pays, personal_historys, titles, registration_dates, deadlines
    ):
        result = {
            "회사명": "".join(company.text.split()),
            "연봉": "".join(salary.text.split()),
            "경력": "".join(history.text.split()),
            "공고제목": "".join(titlee.text.split()),
            "등록일": "".join(r.text.split()),
            "마감일": "".join(d.text.split())
        }
        results.append(result)

# 결과 DataFrame 생성
df = pd.DataFrame(results)

# 결과 출력
print(df.head())

# CSV 저장 (선택)
df.to_csv("work24_ai_jobs_clean.csv", index=False, encoding="utf-8-sig")

# 브라우저 종료
driver.quit()
```

```python
import pandas as pd
import numpy as np
import re

# 1. 환율 설정
EXCHANGE_RATE = 0.0007278  # 2025년 7월 9일 기준

# 2. merge 데이터프레임 예시 불러오기 (이미 있는 경우 생략)
# merge = pd.read_csv("merge.csv")  # 혹은 merge는 이미 존재

# 3. data 컬럼 순서 정의
# 실제 data 기준 컬럼 순서로 정렬
data_columns = [
    "salary_usd", "job_title", "experience_level", "employment_type",
    "company_location", "company_size", "remote_ratio", "required_skills",
    "education_required", "years_experience", "industry"
]

# 맞춰서 재정렬
converted_data = new_aligned[data_columns]

# 4. 빈 DataFrame 생성
new_aligned = pd.DataFrame(np.nan, index=range(len(merge)), columns=columns)

# 5. 채워넣기
new_aligned["job_title"] = merge["공고제목"]
new_aligned["company_name"] = merge["회사명"]

# 연봉 처리: 평균 추출 후 KRW→USD 환산
def parse_salary_krw(s):
    numbers = [int(n.replace(",", "").replace("만원", "")) * 10000
               for n in re.findall(r"[\d,]+만원", str(s))]
    return np.mean(numbers) if numbers else np.nan

def convert_krw_to_usd(krw): 
    return round(krw * EXCHANGE_RATE, 2) if not pd.isna(krw) else np.nan

new_aligned["salary_usd"] = merge["연봉"].apply(parse_salary_krw).apply(convert_krw_to_usd)
new_aligned["salary_currency"] = "KRW"

# 경력 → 연차 수치로 변환
def parse_experience(row):
    if "무관" in row or "신입" in row:
        return 0
    match = re.search(r"(\d+)년", row)
    return int(match.group(1)) if match else np.nan

new_aligned["years_experience"] = merge["경력"].apply(parse_experience)

# experience_level: 0=Entry, 1=Mid, 2=Senior
def level_map(x):
    if pd.isna(x): return np.nan
    if x == 0: return 0
    elif x <= 3: return 1
    else: return 2

new_aligned["experience_level"] = new_aligned["years_experience"].apply(level_map)

# 날짜 처리
new_aligned["posting_date"] = merge["등록일"].str.replace("등록일:", "", regex=False)
new_aligned["application_deadline"] = merge["마감일"].str.replace("마감일:", "", regex=False)

# 인원수
def map_company_size(count):
    if pd.isna(count):
        return np.nan
    if count <= 50:
        return 0
    elif count <= 250:
        return 1
    else:
        return 2

new_aligned["company_size"] = merge["인원수"].apply(map_company_size)

# 나머지 기본값
new_aligned["employment_type"] = 2  # Full-time
new_aligned["company_location"] = 1
new_aligned["employee_residence"] = 1
new_aligned["remote_ratio"] = 0
new_aligned["education_required"] = 0
new_aligned["industry"] = "Unknown"
new_aligned["required_skills"] = np.nan
new_aligned["job_description_length"] = np.nan
new_aligned["benefits_score"] = np.nan

# 6. data 컬럼 기준으로 불필요한 열 제거
new_aligned = new_aligned[data_columns]
```
