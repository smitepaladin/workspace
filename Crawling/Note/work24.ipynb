```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from bs4 import BeautifulSoup
import pandas as pd
import time

driver = webdriver.Chrome()

data =[]
driver = webdriver.Chrome()
driver.get("https://www.work24.go.kr/wk/a/b/1200/retriveDtlEmpSrchList.do?basicSetupYn=&careerTo=&keywordJobCd=&occupation=012&seqNo=&cloDateEndtParam=&payGbn=Y&templateInfo=&rot2WorkYn=&shsyWorkSecd=&resultCnt=10&keywordJobCont=&cert=&moreButtonYn=Y&minPay=1&codeDepth2Info=11000&currentPageNo=1&eventNo=&mode=&major=&resrDutyExcYn=&eodwYn=&sortField=DATE&staArea=&sortOrderBy=DESC&keyword=&termSearchGbn=&carrEssYns=&benefitSrchAndOr=O&occupationParam=012&disableEmpHopeGbn=&actServExcYn=&keywordStaAreaNm=&maxPay=&emailApplyYn=&codeDepth1Info=11000&keywordEtcYn=&regDateStdtParam=&publDutyExcYn=&keywordJobCdSeqNo=&viewType=&exJobsCd=&templateDepthNmInfo=&region=&employGbn=&empTpGbcd=1&computerPreferential=&infaYn=&cloDateStdtParam=&siteClcd=all&searchMode=Y&birthFromYY=&indArea=&careerTypes=&subEmpHopeYn=&tlmgYn=&academicGbn=&templateDepthNoInfo=&foriegn=&entryRoute=&mealOfferClcd=&basicSetupYnChk=&station=&holidayGbn=&srcKeyword=&academicGbnoEdu=noEdu&enterPriseGbn=&cloTermSearchGbn=&birthToYY=&keywordWantedTitle=&stationNm=&benefitGbn=&keywordFlag=&notSrcKeyword=&essCertChk=&depth2SelCode=&keywordBusiNm=&preferentialGbn=&rot3WorkYn=&regDateEndtParam=&pfMatterPreferential=&pageIndex=1&termContractMmcnt=&careerFrom=&laborHrShortYn=#scrollLoc")

# //*[@id="mForm"]/div[2]/div/div[2]/div/div/div/button[3]
# //*[@id="mForm"]/div[2]/div/div[2]/div/div/div/button[4]
#//*[@id="mForm"]/div[2]/div/div[2]/div/div/div/button[12]
# //*[@id="mForm"]/div[2]/div/div[2]/div/div/div/button[5]
for _ in range(1,13):

    for i in range(2,13):  # 2~5 페이지
        xpath = f'//*[@id="mForm"]/div[2]/div/div[2]/div/div/div/button[{i}]'
        element = driver.find_element(By.XPATH, xpath)

        driver.execute_script("arguments[0].scrollIntoView(true);", element)
        time.sleep(0.5)

        driver.execute_script("arguments[0].click();", element)
        time.sleep(1)  # 페이지 로드 대기

        html = driver.page_source
        soup = BeautifulSoup(html, "html.parser")

        # 데이터 추출
        name = soup.select("a.cp_name")
        pay = soup.select("li.dollar span")
        personal_history = soup.select("li.member > p > span:nth-child(1)")
        title = soup.select("td.al_left.pd24 div:nth-child(2) > a")
        registration_date = soup.select("td:nth-child(3) > p:nth-child(4)")
        deadline = soup.select("td:nth-child(3) > p:nth-child(3)")

        # 공백 제거
        company_names = ["".join(a.text.split()) for a in name]
        pays = ["".join(p.text.split()) for p in pay]
        personal_historys = ["".join(p.text.split()) for p in personal_history]
        titles = ["".join(t.text.split()) for t in title]
        registration_dates = ["".join(r.text.split()) for r in registration_date]
        deadlines = ["".join(d.text.split()) for d in deadline]

        # 출력
        for company, salary, history, titlee, r, d in zip(company_names, pays, personal_historys, titles, registration_dates, deadlines):
            print(f"{company} | {salary} | {history} | {titlee} | {r} | {d}")
            data.append([company, salary, history, titlee, r, d])

# 결과 DataFrame 생성
df = pd.DataFrame(data)

# 결과 출력
print(df.head())

# CSV 저장 (선택)
df.to_csv("../Data/교육·법률복지의료예술방송정보통신등전문서비스관리직.csv", index=False, encoding="utf-8-sig")

# 브라우저 종료
driver.quit()
```
