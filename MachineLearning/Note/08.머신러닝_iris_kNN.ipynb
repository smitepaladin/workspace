### Train과 Test Data 분리하기

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
import collections # 빈도수 세기
```

```python
X_train, X_test, y_train, y_test = \
  train_test_split(
    iris_dataset['data'],
    iris_dataset['target'],
    random_state=42,
    stratify=iris_dataset['target'],
    test_size=0.2
  )
```

```python
print(X_train.shape)
print(y_train.shape)
print(X_test.shape)
print(y_test.shape)
```

### kNN으로 학습 하여 정확도 구하기
- n_neighbors = 1
- test의 예측값 출력
- y와 y^의 정확도 출력

```python
from sklearn.neighbors import KNeighborsClassifier

kn = KNeighborsClassifier(n_neighbors=1)
kn.fit(X_train, y_train)
print("Train score:", kn.score(X_train, y_train))
print("Test score:", kn.score(X_test, y_test))
kn.score
```

```python
# 예측값을 구하기
pred = kn.predict(X_test)
pred
```

```python
# 예측값과 실제 test target과 동일한 데이터의 평균이 정확도와 일치하는지 확인
(pred == y_test).mean()
```
