### 랜덤 포레스트(RandomForest)
- 집단학습을 기반으로 고정밀 분류, 회귀, 클러스터링 등을 구현
- 학습 전용 데이터를 기반으로 다수의 의사결정나무를 만들고 만들어진 의사결정나무를 기반으로 다수결과 결과를 유도 한다.

### iris를 RandomForest로 분류

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
import collections # 빈도수 세기
```

```python
iris = pd.read_csv("../Data/iris.csv")
iris.head()
```

```python
# Feature와 Target 분리하기

iris_data = iris.iloc[:,:4]
iris_label = iris.loc[:,'Name']


# iris_data = np.column_stack((iris.PetalLength, iris.PetalWidth))
# iris_label = np.array(iris['Name'])
```

```python
iris_data.head()
```

```python
iris_label.head()
```

```python
print(iris_data.shape)
print(iris_label.shape)
```

```python
# Train과 Test 나누기

train_data, test_data, train_target, test_target = \
  train_test_split(
    iris_data,
    iris_label,
    random_state=42,
    stratify=iris_label,
    test_size=0.2
  )
```

```python
print(train_data.shape)
print(train_target.shape)
print(test_data.shape)
print(test_target.shape)
```

```python
# RandomForest로 학습 시키고 예측하기
from sklearn.ensemble import RandomForestClassifier
clf = RandomForestClassifier()
clf.fit(train_data, train_target)
```

```python
# 예측력 확인
print(clf.score(train_data, train_target))
print(clf.score(test_data, test_target))
```

```python
# 분류 Report
from sklearn import metrics
pred = clf.predict(test_data)
pred
```

```python
clf_report = metrics.classification_report(test_target, pred)
print(clf_report)
```

### 전처리(PreProcessing)이 필요

```python
# 1~22번 컬럼의 Data가 문자로 되어 있어 숫자로 변환
label = []
data = []

for row_index, row in mr.iterrows():
  label.append(row.loc[0])
  row_data = [ord(v) for v in row.loc[1:]]
  data.append(row_data)
```

```python
len(label)
```

```python
len(data)
```

```python
import numpy as np
```

```python
np.array(data).shape
```

```python
import pandas as pd
```

```python
# label을 DataFrame으로 만들자
labelTemp = \
  pd.DataFrame(
    label
  )
labelTemp.head()
```

```python
# data를 DataFrame으로 만들자
dataTemp = \
  pd.DataFrame(
    data
  )
dataTemp.rename(columns=lambda x: x+1, inplace=True) # x값이 들어오면 x+1로 바꾼다. 익명함수
dataTemp.head()
```

```python
# 2개의 DataFrame을 하나로 합치자
mr2 = \
pd.concat(
  [labelTemp,dataTemp],
  ignore_index=True,
  axis=1
)

mr2.head()
```

```python
# Feature와 Target 분리하기

mushroom_data = mr2.iloc[:, 1:]
mushroom_label = mr2.iloc[:, 0]

mushroom_data.head()
```

```python
print(mushroom_data.shape)
print(mushroom_label.shape)
```

```python
# Train과 Test 나누기

train_data, test_data, train_target, test_target = \
  train_test_split(
    mushroom_data,
    mushroom_label,
    random_state=42,
    stratify=mushroom_label,
    test_size=0.2
  )
```

```python
print(train_data.shape)
print(train_target.shape)
print(test_data.shape)
print(test_target.shape)
```

```python
# RandomForest로 학습 시키고 예측하기
from sklearn.ensemble import RandomForestClassifier
clf = RandomForestClassifier()
clf.fit(train_data, train_target)
```

```python
# 예측력 확인
print(clf.score(train_data, train_target))
print(clf.score(test_data, test_target))
```

```python
# 분류 Report
from sklearn import metrics
pred = clf.predict(test_data)
pred
```

```python
clf_report = metrics.classification_report(test_target, pred)
print(clf_report)
```

#### One Hot Encoding
: 숫자 데이터가 숫자로써 의미가 있으면 상관없으나, 위의 데이터는 분류를 위한 데이터 이므로 숫자 크기의 의미가 없다.   
이때 사용하는 것이 One Hot Encoding 이다.

```python
mr2[1].unique()
```

```python
# 연습하기
pd.get_dummies(
  data = dataTemp,
  columns=[1]
)
```

```python
# 적용하기

for i in range(1,22+1):
  dataTemp = pd.get_dummies(
    data = dataTemp,
    columns=[i]
  )
```

```python
mr3 = pd.concat(
  [labelTemp, dataTemp],
  axis = 'columns'
)

mr3.head()
```

```python
# Feature와 Target 분리하기

mushroom_data = mr3.iloc[:, 1:]
mushroom_label = mr3.iloc[:, 0]

mushroom_data.head()
```

```python
print(mushroom_data.shape)
print(mushroom_label.shape)
```

```python
# Train과 Test 나누기

train_data, test_data, train_target, test_target = \
  train_test_split(
    mushroom_data,
    mushroom_label,
    random_state=42,
    stratify=mushroom_label,
    test_size=0.2
  )
```

```python
# RandomForest로 학습 시키고 예측하기
from sklearn.ensemble import RandomForestClassifier
clf = RandomForestClassifier()
clf.fit(train_data, train_target)
```

```python
# 예측력 확인
print(clf.score(train_data, train_target))
print(clf.score(test_data, test_target))
```

```python
# 분류 Report
from sklearn import metrics
pred = clf.predict(test_data)
pred
```

```python
clf_report = metrics.classification_report(test_target, pred)
print(clf_report)
```
