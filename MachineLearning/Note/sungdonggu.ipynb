```python
import pandas as pd
import os
from glob import glob

# 1. 파일 경로 지정
folder_path = "../Data/tpss_bcycl_od_statnhm_202401/"
file_pattern = os.path.join(folder_path, "tpss_bcycl_od_statnhm_202401*.csv")

# 2. glob으로 해당 패턴의 모든 파일 경로 가져오기
csv_files = sorted(glob(file_pattern))

# 3. 데이터프레임 리스트에 각 파일 읽기
df_list = [pd.read_csv(file, encoding='cp949') for file in csv_files]

# 4. 하나의 데이터프레임으로 병합
df_all = pd.concat(df_list, ignore_index=True)

# 5. 확인 및 저장
print(f"총 {len(df_all)}개의 행이 병합되었습니다.")
df_all.to_csv("../Data/tpss_bcycl_od_statnhm_202401_merged.csv", index=False, encoding='utf-8-sig')

```

```python
import pandas as pd
import os
from glob import glob

# 연도 지정
year = 2023

# 1월부터 12월까지 반복
for month in range(1, 13):
    # 월을 두 자리로 포맷팅
    month_str = f"{month:02d}"
    
    # 1. 폴더 및 파일 패턴 정의
    folder_path = f"../Data/tpss_bcycl_od_statnhm_{year}{month_str}/"
    file_pattern = os.path.join(folder_path, f"tpss_bcycl_od_statnhm_{year}{month_str}*.csv")
    
    # 2. 파일 리스트 가져오기
    csv_files = sorted(glob(file_pattern))
    
    # 3. 파일 존재 여부 확인
    if not csv_files:
        print(f"{year}-{month_str} → 해당 월의 CSV 파일이 존재하지 않습니다.")
        continue
    
    # 4. 파일 읽기 (인코딩 오류 대비)
    df_list = []
    for file in csv_files:
        try:
            df = pd.read_csv(file, encoding='cp949')
        except UnicodeDecodeError:
            try:
                df = pd.read_csv(file, encoding='utf-8')
            except UnicodeDecodeError:
                df = pd.read_csv(file, encoding='utf-8-sig')
        df_list.append(df)
    
    # 5. 병합 및 저장
    df_all = pd.concat(df_list, ignore_index=True)
    output_file = f"../Data/tpss_bcycl_od_statnhm_{year}{month_str}_merged.csv"
    df_all.to_csv(output_file, index=False, encoding='utf-8-sig')
    print(f"{year}-{month_str} → {len(df_all)}개 행 병합 후 저장 완료.")
```
