### Data 폴더에 fruits 폴더 만들기

#### Numpy를 Image로 변환하기

```python
import numpy as np
import warnings
warnings.filterwarnings('ignore')
```

```python
fruits = np.load("../Data/fruits.npy")
fruits.shape
```

```python
# Image로 저장하기
from PIL import Image

for i in range(300):
  image = Image.fromarray(fruits[i]).convert('L') # L : Gray
  image.save("../Data/fruits/fruits_%03d.jpg"%i, 'JPEG')
```

----
### Image를 읽어서 Numpy로 변환하기

```python
import glob
```

```python
numberOfData = 300
imgSize = 100
```

```python
# 배열 초기화로 만들기
fruitsData = np.zeros(numberOfData*imgSize*imgSize, dtype=np.int32).reshape(numberOfData, imgSize, imgSize)
fruitsData.shape
```

```python
i = 0
for file in sorted(glob.glob("../Data/fruits/*.jpg")):
  img = np.array(Image.open(file), dtype=np.int32)
  fruitsData[i,:,:] = img
  i+=1
```

```python
fruitsData.shape
```

```python
# 확인
import matplotlib.pyplot as plt

plt.imshow(fruitsData[0].reshape(100,100), cmap='gray_r')

plt.show()
```

```python
np.save("../Data/fruitsData", fruitsData)
```
