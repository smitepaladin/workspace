```python
import pandas as pd
```

```python
df = pd.read_csv("../Data/따릉이_성동구_대여소.csv")
# Initial overview
print(f"Dataset Shape: {df.shape}")
print(f"\nMemory Usage: {df.memory_usage().sum() / 1024**2:.2f} MB")
print("\nFirst few rows:")
df.head()
```

```python
import folium

# 위도, 경도, 대여소 이름 불러오기
latitudes = df['위도']
longitudes = df['경도']
locations = df['주소1']  # 또는 df['주소1']

# 서울 중심 기준 지도 생성
seoul_map = folium.Map(
    location=[37.5665, 126.9780],
    zoom_start=13
)

# 마커 및 원 추가 (기본 아이콘 사용)
for name, lat, lng in zip(locations, latitudes, longitudes):
    popup = folium.Popup(name, max_width=200)
    folium.Marker(
        location=[lat, lng],
        popup=popup,
        icon=folium.Icon(color='blue', icon='bicycle', prefix='fa')  # FontAwesome 자전거 아이콘
    ).add_to(seoul_map)

    folium.CircleMarker(
        location=[lat, lng],
        radius=8,
        color='brown',
        fill_color='coral',
        fill=True,
        fill_opacity=0.6
    ).add_to(seoul_map)

# Jupyter에서는 seoul_map만 호출하면 자동 렌더링
seoul_map
```
