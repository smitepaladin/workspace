```python
from selenium import webdriver
from bs4 import BeautifulSoup
import pandas as pd
import time

# 크롬 드라이버 경로 설정 (또는 webdriver-manager 사용 가능)
driver = webdriver.Chrome()

# 페이지 열기
driver.get("https://www.levels.fyi/companies/google/salaries/software-engineer?country=254")
time.sleep(5)  # 렌더링 대기

```

```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from bs4 import BeautifulSoup
import time



# 5. HTML 파싱
soup = BeautifulSoup(driver.page_source, 'html.parser')

# 6. 레벨 추출 (복수 행 대상)
levels = soup.select("table tbody tr td:nth-child(1) h6")

# 7. 텍스트로 추출
for idx, level in enumerate(levels[:10], 1):  # 앞 10개만 출력 예시
    print(f"{idx}. Level: {level.get_text(strip=True)}")

# 8. 브라우저 종료
driver.quit()

# #company-page_cardContainerId__HLkRd > div > div.MuiBox-root.css-0 > div > div.MuiTableContainer-root.job-family_tableContainer__22xnt.css-kge0eu > table > tbody > tr:nth-child(1) > td:nth-child(2) > h6 연봉
# 
```
