### 군집화(Clustering)
: 비지도학습으로 정답 없이, 특징만 가지고 있는 데이터로 모델을 구현


```python
import pandas as pd
import warnings
warnings.filterwarnings('ignore')
```

```python
df = pd.read_csv("../Data/student_kmeans.csv")
df
```

```python
import matplotlib.pyplot as plt
import seaborn as sns


sns.lmplot(
  x='height',
  y='weight',
  data=df,
  fit_reg=False,

)

plt.show()
```

### k-Means 군집화
몇개구나 k, 평균 Means

```python
from sklearn.cluster import KMeans
```

```python
data_points = df.values

data_points
```

```python
kmeans = KMeans(
  n_clusters=4
).fit(data_points)
```

```python
# 각 군집의 중심위치
kmeans.cluster_centers_
```

```python
kmeans.labels_
```

```python
df['cluster_id'] = kmeans.labels_
df
```

```python
# 시각화

sns.lmplot(
  x='height',
  y='weight',
  data=df,
  fit_reg=False,
  hue='cluster_id'

)

plt.show()
```

```python
# iris를 읽어서 Name 컬럼 제외한 데이터로 3분류 해보기

iris = pd.read_csv("../Data/iris.csv")
iris.head()
```

```python
iris['Name'].unique()
# test_df['label'] = test_df['type'].map({"positive":1, "negative":0})
```

```python
iris_data = iris[['SepalLength','SepalWidth','PetalLength','PetalWidth']]
iris_data.head()
```

```python
data_points = iris_data.values

data_points
```

```python
kmeans = KMeans(
  n_clusters=3
).fit(data_points)
```

```python
kmeans.labels_
```

```python
iris['cluster_id'] = kmeans.labels_
iris.head()
```

```python
iris['label'] = iris['Name'].map({"Iris-setosa":0, "Iris-virginica":1, "Iris-versicolor":2})
iris.tail()
```

```python
len(iris[iris['cluster_id'] == iris['label']])
```

```python
sns.lmplot(
  x='PetalLength',
  y='PetalWidth',
  data=iris,
  fit_reg=False,
  hue='label'

)

plt.show()
```
