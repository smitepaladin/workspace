#### 머신러닝 이란
: 수많은 데이터를 학습시켜 거기에 있는 패턴(=경향)을 찾아내고 이를 기반으로 데이터를 분류(Classification)하거나 예측(predict).
- 일반 소프트웨어(결정론적 소프트웨어)는 언제나 똑같은 과정을 거쳐 똑같은 결과를 냅니다.
- 머신러닝 기반 소프트웨어는 데이터와 알고리즘에 따라 때로는 다른 결과를 추론해 냅니다.

추론 : 이미 알고있는 것으로 논리적 결론을 도출하는 과정
- 머신러닝(비결정론적 소프트웨어) : 데이터와 알고리즘을 기반으로 추론하는 프로그램으로 정의한다.

##### 결정론적 소프트웨어와 비결정론적 소프트웨어의 차이점
: 예를 들어 지문인식기의 소프트웨어를 결정론적 소프트웨어로 처리하였다고 하면 지문 입력기에 지문이 조금만 틀어져도 거부의 메세지가 나타난다.   
이에 반해 비결정론적 소프트웨어는 지문이 조금 틀어져도 추론을 통해 지문의 가부를 결정한다.   

#### 머신러닝 개요
: 머신러닝(Machine Learning)이란 인공지능 연구과제 중의 하나로 인간의 뇌가 자연스럽게 수행하는 '학습'이라는 능력을 컴퓨터로 구현하는 방법   
- 어느정도 규모가 있는 샘플데이터를 입력으로 넣어 분석 시킵니다.
- 그리고 이때 데이터에서 일정한 규칙을 찾아 냅니다.
- 그리고 찾아낸 규칙을 기반으로 다른 데이터를 분류하거나 예측하는 것

##### 특징(Feature, Attribute) 추출
특징 추출이란 머신러닝을 하기 위해 데이터가 어떤 특징을 가지고 있는지 찾고 벡터로 만드는 처리과정(DP만들기)     
일반적으로 사용되는 머신러닝 프레임워크는 범용적인 용도입니다.    
따라서 내가 원하는 특징을 추출해 주는 기능은 없습니다.   
어떤 특징을 추출 할지는 분석가가 결정해야 합니다.    
이것이 머신러닝의 중요한 포인트 입니다.

#### 회귀분석
회귀분석(regression analysis)은 Y가 연속된 값일 때 Y=f(x)와 같은 모델로 나타내는 것 입니다.    
회귀에서 사용되는 가장 기본적인 모델은 Y = aX + b이고 이를 선형회귀라고 부릅니다.    
이때 Y를 종속변수(Target), X를 독립변수(Feature, Input)라고 부릅니다.     
X가 1차원이고 한개이면 단순회귀    
X가 1차원이고 여러개이면 다항회귀    
X가 2차원 이상이면 다중회귀   

### 머신러닝의 종류
##### 지도학습(Supervised Learning)
: 학습시 데이터와 답을 같이 학습한 후, 다른 데이터의 답을 예측한다.    
##### 비지도학습(Unsupervised Learning)
: 학습시 데이터만 학습한 후, 다른 데이터의 규칙성을 찾을 때. 본질적인 구조등을 확인할 때 사용     
##### 강화학습(Reinforcement Learning)
: 부분적으로 답을 입력한 후, 데이터를 기반으로 최적의 답을 찾아 냅니다.   
지도학습과 비슷하지만 완전한 답을 제공하지 않는것이 차이점 입니다.

## 머신러닝의 흐름 
(1) 데이터 수집    
(2) 데이터 가공 : 정제, 전처리    
(3) 데이터 학습   
(3-1) 학습 방법 선택    
(3-2) 매개변수(Hyper Parameter) 조정    
(3-3) 모델 학습   
(4) 모델 평가   
(5) 예측력이 좋은 경우 다음 단계를 진행하고 아니면 (3) 재실행    
(6) 성공   


#### 머신러닝의 응용 분야
##### 클래스 분류(Classification)
- 특정 데이터에 레이블을 붙여 분류할 수 있습니다.
- 스팸메일 분류, 필기 인식, 증권 사기 등   


##### 클러스터링(Clustrering) : 그룹 나누기   
- 값의 유사성을 기반으로 데이터를 여러 그룹으로 나눌 수 있다.
- 사용자의 취향을 그룹으로 묶어 사용자 취향에 맞는 광고를 제공.    


##### 추천(Recommendation)
- 특정 데이터를 기반으로 다른 데이터를 추천하는 것 입니다.   
- 인터넷 서점에서 구매한 책들을 기반으로 다른 책을 추천.   


##### 회귀(Regression)
- 과거의 데이터를 기반으로 미래를 예측하는 것 입니다.   
- 판매예측, 주가변동   


##### 차원축소(Dimensionality Reduction)
- 데이터의 특성을 유지하면서 데이터의 양을 줄이는 것    
- 데이터를 시각화 하거나 구조를 추출해서 용량을 줄여 계산을 빠르게 하거나 메모리를 절약할 때 사용.

### 프로젝트 과정
: 머신러닝 프로젝트는 크게 문제정의, 데이터 획득, 모델 구현, 테스트의 단계로 구현    



##### 문제정의
- 프로젝트의 목표를 정의하고 그에 따른 계획을 세우는 단계
- 대략적으로 어떤 데이터를 수집해서 어떤 머신러닝 알고리즘으로 문제를 해결할지 결정하는 단계   



##### 데이터 획득
- 머신러닝 모델 학습에 필요한 데이터를 수집하는 단계
- 성공적인 머신러닝 프로젝트는 뛰어난 머신러닝 실력 보다도 얼마나 충분한 데이터를 획득했느냐에 의해 결정
- 획득한 데이터는 학습(Training) 데이터, 검증(Valid) 데이터, 테스트(Test) 데이터로 구분하여 사용됩니다.   


##### 모델 구현
- 학습 데이터를 기반으로 한개 이상의 머신러닝 모델을 구현


##### 검증
- 검증 데이터를 사용해 구현된 머신러닝 모델들의 성능을 검증
- 최종적으로 검증결과가 가장 높은 모델 하나를 선택


##### 테스트
- 검증 결과를 통해 선택된 모델의 성능을 테스트 데이터로 측정
- 테스트 결과에 따라 프로젝트를 마무리 할지, 아니면 이전 단계로 돌아가서 개선할지를 결정
