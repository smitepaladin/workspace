#### 교차검증과 그리드 서치
- 머신러닝을 사용할 때 모델의 정확도를 측정하기 위해 반드시 사용해야 하는 방법
- 딥러닝에서는 데이터의 크기가 크므로 이 방법은 사용할 필요가 없다.

```python
import pandas as pd
import warnings
warnings.filterwarnings('ignore')
```

```python
wine = pd.read_csv("../Data/wine.csv")
wine.head()
```

```python
# Feature와 Target
data = wine.iloc[:,:3]
target = wine.iloc[:,3]
```

### Train, Valid, Test로 분리

```python
from sklearn.model_selection import train_test_split

train_input, test_input, train_target, test_target = \
  train_test_split(
    data,
    target,
    test_size=0.2,
    random_state=42
  )
```

```python
sub_input, val_input, sub_target, val_target = \
  train_test_split(
    train_input,
    train_target,
    test_size=0.2,
    random_state=42
  )
```

```python
# 크기 확인

print("Train :", sub_input.shape)
print("Valid :", val_input.shape)
print("Test  :", test_input.shape)
```

```python
# 훈련세트와 검증세트를 결정트리로 모델 만들기
from sklearn.tree import DecisionTreeClassifier

dt = DecisionTreeClassifier(random_state=42)
dt.fit(sub_input, sub_target)

print("Train :", dt.score(sub_input, sub_target))
print("Valid :", dt.score(val_input, val_target))
```

```python
### 교차검증
```

```python
from sklearn.model_selection import cross_validate
scores = cross_validate(dt, train_input, train_target)
scores
```

```python
# 교차검증 후의 정확도
import numpy as np

np.mean(scores['test_score'])
```

### KFold를 이용한 방법 : 교차검증

```python
from sklearn.model_selection import StratifiedKFold

splitter = StratifiedKFold() # 타겟 기준으로 비율 맞춰서 하겠다.
scores = cross_validate(dt, train_input, train_target, cv=splitter) # KFold는 단독으로 사용하지 않는다. 
scores
```

```python
np.mean(scores['test_score'])
```

```python
# KFold 중 fold를 10개로
splitter = StratifiedKFold(n_splits=10, shuffle=True, random_state=42) # 타겟 기준으로 비율 맞춰서 하겠다.
scores = cross_validate(dt, train_input, train_target, cv=splitter) # KFold는 단독으로 사용하지 않는다. 
scores
```

```python
np.mean(scores['test_score'])
```

```python
# KFold 중 fold를 10개로
splitter = StratifiedKFold(n_splits=10, shuffle=False) # 타겟 기준으로 비율 맞춰서 하겠다.
scores = cross_validate(dt, train_input, train_target, cv=splitter) # KFold는 단독으로 사용하지 않는다. 
scores
```

```python
np.mean(scores['test_score'])
```

---
### 그리드 서치(Grid Search)를 이용한 최적의 Hyper parameter 값 찾기

```python
from sklearn.model_selection import GridSearchCV

params = {'min_impurity_decrease' : [0.0001, 0.0002, 0.0003, 0.0004, 0.0005]}
```

```python
gs = GridSearchCV(
  DecisionTreeClassifier(random_state=42),
  params,
  n_jobs=-1
)
```

```python
gs.fit(train_input, train_target)
```

```python
dt = gs.best_estimator_
print(dt.score(train_input, train_target))
```

```python
gs.best_params_
```

```python
gs.cv_results_['mean_test_score'].mean()
```

### Random Search
: 정해진 파라미터가 아니고 파라미터의 범위를 정해 최적의 값 찾기

```python
# 확률 분포 중 균등 분포
from scipy.stats import uniform, randint
```

```python
params = {
  'min_impurity_decrease' : uniform(0.00001, 0.0001),
  'max_depth' : randint(20,50),
  'min_samples_split' : randint(2,25),
  'min_samples_leaf' : randint(1,25)
}
```

```python
from sklearn.model_selection import RandomizedSearchCV

gs = RandomizedSearchCV(
  DecisionTreeClassifier(random_state=42),
  params,
  n_iter=100,
  n_jobs=-1,
  random_state=42
)
```

```python
gs.fit(train_input, train_target)
```

```python
gs.best_params_
```

```python
np.mean(gs.cv_results_['mean_test_score'])
```

```python
dt = gs.best_estimator_
print(dt.score(train_input, train_target))
print(dt.score(test_input, test_target))
```

```python
print("Train :", dt.score(sub_input, sub_target))
print("Valid :", dt.score(val_input, val_target))
print("Test  :", dt.score(test_input, test_target))
```
