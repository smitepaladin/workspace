## 지도학습과 비지도 학습
- 학습 : 데이터를 특별한 알고리즘에 적용해 머신러닝 모델을 정의된 문제에 최적화하는 과정을 의미

##### 지도학습(supervised learning)
- 정답을 알려주면서 진행되는 학습
- 정답, 실제값, 레이블, 타깃, 클래스, y값 <- 동일한 의미
- 머신러닝 모델을 통해 예측된 값을 예측값, 분류값, y hat(예측값)등으로 표현
- 지도학습의 예로는 분류(Classification)와 회귀(Regression)이 대표적

##### 비지도학습(unsupervised learning)
- 정답(레이블)이 없이 진행되는 학습
- 데이터 자체에서 패턴을 찾아내야 할 때 사용
- 비지도 학습의 예로는 군집화, 차원축소가 대표적

----
#### 분류와 회귀
- 분류와 회귀의 큰 차이점은 데이터가 입력 되었을 때 분류는 분리된 값(객관식)으로 예측하고 회귀는 연속된 값(주관식)으로 예측하는데 있다.

##### 분류(Classification)
: 분류는 데이터가 입력되었을 때 지도학습을 통해 미리 학습된 레이블 중 하나 또는 여러개의 레이블로 예측하는 것.

###### 이진분류
- 둘 중 하나의 값으로 분류하는 경우
###### 다중분류
- 여러개의 분류값 중에서 하나의 값으로 예측하는 경우

#### 회귀
- 입력된 데이터에 대해 연속된 값으로 예측
- 분류는 정해진 레이블을 예측하는 반면 회귀는 연속성을 가진 수치로 예측

----
### 과대적합(overfitting)과 과소적합(underfitting)
- 데이터에서 충분히 특징을 찾아내지 못하고 머신러닝 모델을 학습할 경우 모델이 과소적합이 되기 쉽고    
- 필요 이상의 특징으로 학습할 경우 모델이 과대적합이 되기 쉽다.   


#### 과소적합
- 보통 테스트 데이터 뿐만 아니라 학습 데이터에 대해서도 정확도가 낮게 나올 경우 과소적합된 모델일 가능성이 높다.   


#### 과대적합
- 학습 데이터외의 데이터에는 정확도가 낮게 나오는 모델
- 데이터의 특지들의 수치값을 정규화함으로써 특성에 의한 편향(bias)을 줄이는 것도 과대적합을 피하는 좋은 방법이다.   
- 딥러닝 같은 경우 조기종료(early stopping), 드롭아웃(drop out)을 사용해 과대적합을 피할 수 있다.

### 혼동행렬(confusion matrix)
- 모델의 성능을 평가할 때 사용되는 지표
- 실제값과 예측값을 비교하여 실제값에 예측값이 어느정도 일치되었는지를 파악
- 모델의 성능은 이 혼동행렬을 기반으로 단 하나의 수치로 표현 가능

<img src="../Data/정확도.png" style="height:80%;" />

- A Class인 경우 : 10번의 예측값 중 9번은 A로 맞추고 1번은 B라고 맞춤
- D Class인 경우 : 20번의 예측값 중 15번은 D로 4번은 1번은 C로 맞춤

#### 머신러닝 모델이 성능 평가
##### TP(True Positive) : 맞는 것을 올바르게 예측한 것
: 9, 15, 24, 15
##### TN(True Negative) : 틀린것을 올바르게 예측한 것
- A class인 경우 : A가 아닌것들을 A가 아니라고 예측한 것
##### FP(False Positive) : 틀린것을 맞다고 잘못 예측한 것
##### FN(False Negative) : 맞는것을 틀렸다고 잘못 예측한 것
##### 정확도(Accuracy)
- 가장 일반적인 모델 성능 평가 지표
- 혼동행렬 상에서는 TP를 전체 셀로 나눈 값에 해당

```python
print(f'정확도 : {(9+15+24+15)/80*100}%')
```

### 정밀도(Precision)

<img src="../Data/정밀도-2.png" style="height:20%; width:10;" />
<img src="../Data/정밀도-1.png" style="height:20%; width:10;" />

#### 2가지 모델의 정확도
- A Model : 69/ 100 -> 69%
- B Model : 71/ 100 -> 71%

```python
# A Model의 정밀도 : TP / (TP + FP)
print(f"A Model 암환자의 정밀도 : {9 / (9+30)}")
```

```python
# B Model의 정밀도 : TP / (TP + FP)
print(f"B Model 암환자의 정밀도 : {1 / (1+20)}")
```

#### 재현율(Recall)
- 실제값 중에서 모델이 검출한 실제값의 비율을 나타내는 지료
- TP / (TP+FN)
- A Model의 암환자 재현율 : 9 / (9+1) = 0.9
- B Model의 암환자 재현율 : 1 / (1+9) = 0.1

#### F1점수(F1 Score)
- 정밀도도 중요하고 재현율도 중요한데 어떤 것을 선택할지 혼돈시 이 두값의 조화평균 내서 하나의 수치로 나타낸 지료를 
- 조화평균 = 2*a*b / (a+b)
- F1점수 = 2 * 재현율 * 정밀도 / (재현율+정밀도)

```python
# A Model의 F1점수 : F1점수 = 2 * 재현율 * 정밀도 / (재현율+정밀도)
print(f"A Model F1점수 : {2 * 0.9 * 0.23 / (0.9 + 0.23)}")
```

```python
# B Model의 F1점수 : F1점수 = 2 * 재현율 * 정밀도 / (재현율+정밀도)
print(f"B Model F1점수 : {2 * 0.1 * 0.04 / (0.1 + 0.04)}")
```

### k폴드 교차 검증(k-fold cross validation)
: 모든 학습 데이터를 한번씩 검증 데이터로 활용해 검증 데이터가 한쪽 데이터에 편향되어 있지 않게 하여 검증 데이터를 분리하지 않고도 학습 데이터에 전반적인 검증 정확도를 구할 수 있음
