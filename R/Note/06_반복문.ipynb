```r
excel <- read.csv("../Data/excel.csv")
head(excel)
```

```r
# Ex : for문을 사용하여 excel의 q3컬럼 데이터의 홀수의 합과 짝수의 합을 구하기
# 출력 예시 : 홀수의 합은 __ 이고 짝수의 합은 __ 입니다.
aaa <- 0
bbb <- 0

for(i in excel$q3){
  if(i %% 2 == 1){
    aaa <- aaa + i
  }else{
    bbb <- bbb + i
  }
}

cat("홀수의 합은", aaa, "이고 짝수의 합은", bbb, "입니다.")
```

```r
aaa <- sum(excel$q3[excel$q3 %% 2 == 1])
bbb <- sum(excel$q3[excel$q3 %% 2 == 0])
cat("홀수의 합은", aaa, "이고 짝수의 합은", bbb, "입니다.")
```
