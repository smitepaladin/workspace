#### 2개 이상의 데이터 프레임 병합하기

```r
# case 1 : Key값이 되는 컬럼이 동일하고 data의 갯수도 동일한 경우
height <- data.frame(
  id=1:3,
  height=c(160,170,180)
)
height
```

```r
weight <- data.frame(
  id=c(2,1,3),
  weight=c(70,60,80)
)
weight
```

```r
# 병합하기
merge(height,weight)
```

```r
# Case 2: Key 값이 동일하지 않고 각각의 data갯수도 다를 경우
height
```

```r
weight = data.frame(
  ad = c(2,1),
  weight = c(70,60)
)
weight
```

```r
merge(height,weight)
```

```r
merge(
  x=height,
  y=weight,
  by.x = 'id',
  by.y = 'ad'
)
```

```r
merge(
  x=height,
  y=weight,
  by.x = 'id',
  by.y = 'ad',
  all.x = T
)
```

#### 문자열 처리
- 수집한 데이터의 필요한 문자열을 자르고, 교체하고, 추출하는 작업

##### stringr package 사용
- str_length(): 문자열 길이
- str_c() : 문자열 연결
- str_sub() : 범이ㅜ에 해당하는 문자열 추출
- str_split() : 구분자를 기준으로 문자열 분리
- str_replace() : 기존 문자열을 특정 문자열로 변환
- str_extract() : 문자열에서 특정 문자열 패턴의 첫번째 문자열 추출
- str_extract_all() : " 모든것 추출
- str_locate() : 문자열에서 문자열 패턴의 첫번째 위치 추출
- str_locate_all() : " 모든 위치 추출

```r
install.packages("stringr")
```

```r
# 패키지 불러오기
library(stringr)
```

```r
string <- 'hongkildong105yoube1002kwanwoo25장비2005'
```

```r
# a~z까지의 영문자가 연속으로 3개 발견된 문자열 추출
str_extract_all(string, "[a-z]{3}")
```

```r
# a~z까지의 영문자가 연속으로 3~5개 발견된 문자열 추출
str_extract_all(string, "[a-z]{3,5}")
```

```r
# 영문자만 가져오기
aaa <- str_extract_all(string, "[a-z]{1,}")
aaa
class(aaa)

# list를 해제

bbb <- unlist(aaa)
bbb
```

```r
# 숫자만 가지오기
ccc <- str_extract_all(string, "[0-9]{1,}")
ccc
```

```r
# 한글만 가지오기
ddd <- str_extract_all(string, "[가-힣]{1,}")
ddd
```

```r
# 영문자를 제외한 문자 추출하기
str_extract_all(string, "[^a-z]{1,}")
```

```r
# 문자열 길이
str_length(string)
```

```r
# 문자열 위치
str_locate(string, "장비")
```

```r
# 부분 문자열
str_sub(string, 1, 32)
```

```r
str_sub(string, 1, str_locate(string, "장비")[1]-1)
```

```r
# 대문자, 소문자 변경
str_to_upper(string)
```

```r
# Ex : string에서 숫자만 추출하여 합계 구하기
sum(as.numeric(unlist(str_extract_all(string, "[0-9]{1,}"))))
```

```r
# 문자열 교체하기
string <- str_replace(string, 'hongkildong105', '홍길동35,')
string <- str_replace(string, 'yoube1002', '유비45,')
string <- str_replace(string, 'kwanwoo25', '관우25,')
string <- str_replace(string, '2005', '20')
string
```

```r
# 문자열 분리
unlist(str_split(string,","))
```

```r
# 천단위 쉼표가 들어간 ㅁ눈자열을 숫자로 변경하기
price <- "123,456,789"

as.numeric(str_replace_all(price,",",""))
```

```r
price <- "123,456,789"
as.numeric(str_remove_all(price,","))
```

```r
# NA 변경하기
temp <- c(1,2,NA,4,NA)
# NA -> 9
as.numeric(str_replace_na(temp,'9'))
```
