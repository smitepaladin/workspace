```python
def loadSpeed_zip(dfName, folderName: str):
    import zipfile
    import glob
    import os
    import warnings
    warnings.filterwarnings('ignore')
    dfName = pd.DataFrame(
        columns=["해당일","시간대","압구정로11길","잠원로상단","잠원로하단"]
    )
    for a in [5,6]:
      # 폴더 안 zip 파일 목록 가져오기
      zip_files = glob.glob(f"{folderName}/{a}/*.zip")
      zip_files.sort()
      
      loads = {'압구정로11길':'1220321100','잠원로상단':'1210030300','잠원로하단':'1210029100'}

      
      for zip_file in zip_files:
          # 1️⃣ zip 하나 풀기
          with zipfile.ZipFile(zip_file, 'r') as zf:
              zf.extractall(folderName)
          print(f"{zip_file} 압축 해제 완료")
          
          # 2️⃣ 방금 풀린 CSV 파일만 가져오기
          csv_files = [os.path.join(folderName, f) for f in zf.namelist() if f.endswith('.csv')][0]
          
          
              # CSV 읽기
          data = pd.read_csv(csv_files, encoding="cp949", header=None)
          data.rename(columns={
                  0:"해당일",
                  1:"시분",
                  2:"LinkID",
                  3:"관리기관",
                  4:"통행속도",
                  5:"통행시간"
              }, inplace=True)
          if data.LinkID.dtype == 'float64':
              data.LinkID = data.LinkID.astype(int).astype(str)
          if data.LinkID.dtype == 'int64':
              data.LinkID = data.LinkID.astype(str)
          df1 = pd.DataFrame(columns=["해당일","시간대","압구정로11길","잠원로상단","잠원로하단"])
          result = pd.DataFrame()
              
          for i in list(loads.keys()):
                  df = data[data.LinkID == loads[i]]
                  df['해당일'] = df['해당일'].astype(str)
                  df['시분_str'] = df['시분'].apply(
                  lambda x: f"{int(x):04d}" if pd.notnull(x) else None).astype(str)
                  df['hour'] = df['시분_str'].str[:2].astype(int)
                  # 시간대별 평균 계산
                  result = df.groupby(['LinkID', "해당일","hour"])['통행속도'].mean().reset_index()
                  df1[i] = result.통행속도
              
          df1.해당일 = result.해당일
          df1.시간대 = result.hour
          dfName = pd.concat([dfName, df1], ignore_index=True)
              
              # 3️⃣ CSV 삭제
          os.remove(csv_files)
          print(f"{csv_files} 삭제 완료")
      dfName.to_csv(f"/Volumes/MacSSD - 데이터/Users/tj/Documents/Development/down2024/2024_{a}월.csv",index=False)
      print(f"{a}월 csv파일 저장 완료")
      
```
