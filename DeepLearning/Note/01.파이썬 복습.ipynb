### indexing과 slicing
: 배열에서 원하는 값 선택, Masking, 조건처리

```python
my_arr
```

```python
print(my_arr[0]) # 0번 행이 나옴
print(my_arr[1])
print(my_arr[1:]) # : 은 차원이 들어간다. 속성값이 들어가는 것이다.
print(my_arr[0:1]) # 1 전까지 차원을 늘려라
print(my_arr[:,1]) # 범위를 쓰지 않았다. 차원을 가져오지 않는다.
print(my_arr[1:,::2]) # step 처음부터 2칸 증가
my_arr[my_arr > 40]
```

### BroadCasting
: 서로 다른 차원의 배열간 연산을 가능하게 함
- 두 배열의 차원수를 오른쪽(행, 열)에서 부터 비교
 - 각 차원의 크기가 일치하거나, 열이 1이면(1차원데이터) 연산 가능하고 그렇지 않으면 오류 발생

```python
# 각 차원의 크기가 일치하는 덧셈
arr1 = np.array(
  [
    [1,1,1],
    [2,2,2],
    [3,3,3]
  ]
)
arr1
```

```python
# 각 차원의 크기가 일치하는 덧셈
arr2 = np.array(
  [
    [10,10,10],
    [20,20,20],
    [30,30,30]
  ]
)
arr2
```

```python
arr1 + arr2
```

```python
# 각 차원의 크기가 일치하지 않거나 각 차원 중 한개의 차원이 1인 경우
arr1 = np.array(
  [
    [1],
    [2],
    [3]
  ]
)

print(arr1)
print(arr1.shape)
```

```python
arr2 = np.array(
    [10,20,30]
)
print(arr2)
print(arr2.shape)
```

```python
print(arr1+arr2)
print(arr2+arr1)
```

```python
# 안되는 경우
arr1 = np.array(
  [
    [1,2,3],
    [4,5,6]
  ]
)


arr2 = np.array(
    [10,20]
)

try: 
  arr1 + arr2
except ValueError as e:
  print("Error :", e)
```

### Map

```python
numbers = [1,2,3,4,5]
# 각 요소를 제공하는 함수
def square(n):
  return n**2

list(map(square, numbers))
```

```python
numbers = [1,2,3,4,5]
list(map(lambda n: n**2, numbers))
```

```python
# 여러 리스트를 동시에 변환
# 결과 = [5,7,9]

list1 = [1,2,3]
list2 = [4,5,6]

list(map(lambda n1,n2: n1+n2, list1,list2))
```

```python
numbers = [1,2,3,4,5]
# 결과 ['홀수', '짝수', ...]

list(map(lambda n: "짝수" if n % 2 == 0 else "홀수", numbers))
```
