### 내장함수

```python
# 절대값
print(abs(-3))
```

```python
# all
print(all([1,2,3]))
print(all([1,2,0]))
```

```python
# any
print(any([0,0,1]))
print(any([0,0,0]))
print(any([0,"",""]))
```

```python
# chr : ASCII Code값을 문자로 출력
print(chr(97))
print(chr(48))
```

```python
# ord : 문자를 ASCII Code로 출력
print(ord("a"))
print(ord("0"))
```

```python
# enumerate : 순서가 있는 자료형을 index가 있는 enumerate 객체로 변환
for i, name in enumerate(['a','b','c']):
  print(i, name)
```

```python
# 양수만 고르는 함수
def positive(numberList):
  result = []
  for num in numberList:
    if num> 0:
      result.append(num)
  return result
```

```python
print(positive([1,-3,0,2,-5,6]))
```

```python
# 양수만 고르는 함수를 filter를 사용
def positive(x):
  return x > 0

print(list(filter(positive, [1,-3,0,2,-5,6]))) # True인것만 가지고 list를 만들겠다
```

```python
var = divmod(7,3) # 몫과 나머지
print(var)
```

```python
# eval(expression)은 실행 가능한 문자열을 입력받아 실행
print(eval("1+2"))
print(eval("'hi'+'data'"))
print(eval("divmod(4,3)"))
```

```python
a1 = [6,3,9]
a2 = a1.sort() # sort는 리턴이 없다.
print(a2)
```

```python
a1 = [6,3,0]
a2 = sorted(a1) # sorted는 리턴이 있다.
print(a2)
```

```python
# zip 위도경도에 많이 쓴다.
num1List = [1,2,3]
num2List = [4,5,6]

for num1, num2 in zip(num1List, num2List):
  print(num1, num2)
```
