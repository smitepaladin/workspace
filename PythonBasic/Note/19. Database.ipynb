## SQlite

```python
import sqlite3

# sqlite 데이터베이스 연결하기

dbPath = "../Data/test.sqlite"
conn = sqlite3.connect(dbPath)


# 테이블을 생성하고 데이터 넣기
curs = conn.cursor()
curs.executescript(
  """
  DROP TABLE IF EXISTS items;

  CREATE TABLE items(
    item_id INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT UNIQUE,
    price INTEGER
  );

  INSERT INTO items(name, price) VALUES ('Apple', 800);
  INSERT INTO items(name, price) VALUES ('Orange', 780);
  INSERT INTO items(name, price) VALUES ('Banana', 430);
  """
)

conn.commit()
```

```python
# 데이터 추출하기
curs = conn.cursor()
curs.execute("SELECT item_id, name, price FROM items")
item_list = curs.fetchall()

for it in item_list:
  print(it)
```

```python
dbPath = "../Data/test.sqlite2"
conn = sqlite3.connect(dbPath)


# 테이블을 생성하고 데이터 넣기
curs = conn.cursor()
curs.executescript(
  """
  DROP TABLE IF EXISTS items;

  CREATE TABLE items(
    item_id INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT UNIQUE,
    price INTEGER
  );
  """
)

conn.commit()
```

```python
# 데이터 넣기

curs = conn.cursor()
curs.execute("INSERT INTO items (name, price) VALUES (?,?)", ("Orange", 5200))
conn.commit()
```

```python
# 여러 데이터를 연속으로 넣기
# 데이터 넣기

curs = conn.cursor()
data =[("Mango", 7700), ('Kiwi', 4000), ('Grape', 8700),('Peach', 9400)]
curs.executemany("INSERT INTO items (name, price) VALUES (?,?)", data)
conn.commit()
```

```python
# 데이터 추출하기
curs = conn.cursor()
curs.execute("SELECT item_id, name, price FROM items")
item_list = curs.fetchall()

for it in item_list:
  print(it)
```

### TinyDB
- NoSQL 문서형 데이터 베이스
- MongoDB가 대표적이지만 MongoDB는 설치가 필요하다

```python
#!pip install tinydb
```

```python
from tinydb import TinyDB, Query, where

# 데이터 베이스 연결
filePath = "../Data/tinydb.json"
db = TinyDB(filePath)
```

```python
# 기존에 테이블이 있다면 제거
db.drop_table('fruits')
```

```python
# 테이블 생성하기
table = db.table('fruits')
```

```python
# 데이터 추가하기
table.insert(
  {
    'name' : 'Banana',
    'price' : 6000
  }
)
table.insert(
  {
    'name' : 'Orange',
    'price' : 12000
  }
)
table.insert(
  {
    'name' : 'Mango',
    'price' : 9000
  }
)
```

```python
# 모든 데이터 추출
print(table.all())
```

```python
# 검색
item = Query()
res = table.search(item.name == 'Orange')
print(res[0]['price'])
```

```python
# 가격이 8000원 이상인 것
res = table.search(item.price >= 8000)
res
```

```python
# 수정하기
table.upsert({'price':8000}, item.name == 'Mango')
print(table.all())
```

```python
# 삭제하기
table.remove(where('name') == 'Mango')
print(table.all())
```
