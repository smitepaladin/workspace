### 자료형의 참과 거짓

- 문자열인 경우 : ""안에 data가 없으면 거짓, data가 있으면 참
- 리스트인 경우 : [] ---> 
- 튜플인 경우 : () ---> 
- 딕셔너리인 경우 : {} --->
- 숫자인 경우 : 0이면 거짓, 0이 아닌 숫자는 참

```python
var = [1,2,3,4]
while(var): # var에 데이터가 있으니 true
  print(var.pop())
var
```

```python
# 튜플을 이용한 변수 생성
var01, var02 = 1, 2
print(var01, var02)
```

```python
var01 = var02 = 1
```

```python
# 변수 값 교환

num1 = 10
num2 = 20

num1, num2 = num2, num1

print(num1, num2)
```

### 자료형 변환

```python
# 문자열을 정수로 변환하기
str_num = '123'
int_num = int(str_num)
print(int_num)
```

```python
# 문자열을 실수로 변환하기
str_num = "123.456"
float_num = float(str_num)
print(float_num)
```

```python
# 정수를 문자로 변환
int_num = 123
str_num = str(int_num)
str_num
```

```python
# 정수를 실수로 변환
int_num = 123
float_num = float(int_num)
float_num
```

```python
# 실수를 정수로 변환
float_num = 123.956
int_num = int(float_num)
int_num
```
