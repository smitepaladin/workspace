### while

```python
# 10번 직어 안 넘어가는 나무 없다.

treeHit = 0

while treeHit < 10:
  treeHit += 1
  print(f"나무를 {treeHit}번 찍었습니다.")
  if treeHit == 10:
    print("나무 넘어 갑니다.")
```

```python
# break
var = 0

while var < 10:
  print(var)
  var += 1
  if var == 5:
    break
```

```python
# continue

var = 0

while var < 10:
  var += 1
  if var == 5:
    continue
  print(var)
```

## Ex
continue문을 사용하여 1에서 10까지의 수 중 홀수만 출력하기

```python
var = 0

while var < 10:
  var += 1
  if var % 2 == 0:
    continue
  print(var)
```

### Ex
- 5번만 커피를 제공하는 자판기 입니다.
- 금액은 300원 입니다.
- 사용자가 300원보다 큰 금액을 입력하면 잔돈을 줍니다.
- 사용자가 300원 보다 작은 금액을 입력하면 돈은 돌려주고 커피를 주지 않으며 남은 커피의 횟수를 보여 줍니다.
- 5번만 커피를 다 팔았을 경우에는 판매를 중지한다는 메시지를 줍니다.

```python
stock = 5
price = 300
```

```python
while stock :
  money = int(input("금액을 입력해주세요"))
  if money >= price:
    stock -= 1
    print(f"커피가 나왔습니다. 잔돈은 {money - price}원 입니다.")
  elif money < price | money >= 0:
    print(f"금액이 부족합니다. 남은 커피갯수는 {stock}개 입니다.")
  else:
    print("올바른 금액을 투입하세요.")
print("품절입니다. 판매를 중지합니다.")

```

```python
# 평균 구하기
scores = [78,85,95,65]
total = 0
list_length = len(scores)

while scores:
  total += scores.pop()

print(total / list_length)
```

```python
# 표준편차 (평균과 각 data의 차이의 제곱한 것의 제곱근을 구한 것의 평균)
scores = [78,85,95,65]

# 평균 구하기
total = 0
var = 0

while var < len(scores):
  total += scores[var]
  var += 1

avg = total / len(scores)
print(avg)

# 각 데이터의 평균의 차이의 제곱의 합의 평균 구하기

total = 0
var = 0

while var < len(scores):
  total += (scores[var] - avg) ** 2
  var += 1

avg = total / len(scores)
print(avg)

import math
math.sqrt(avg)
```

### while else
: else문은 반복분이 정상적으로 종료 되었을 경우 실행된다.   
: while 문 안에 break가 있어 반복이 중단된 경우에는 else문을 실행하지 않는다.

```python
i = 1
while i < 5:
  print(i)
  if i == 3:
    break
  i += 1
else:
  print('반복이 종료 되었습니다.')
```
