### 관계 연산자

```python
num1 = 100
num2 = 20


# 동등비교
print(num1 == num2)
print(num1 != num2)

# 크기비교
print(num1 > num2)
print(num1 < num2)
print(num1 >= num2)
print(num1 <= num2)
```

### 논리연산자

```python
# And, Or 연산자 (&, |, <= 사용하면 안됨, bit연산자)
print((num1 >= 50 and (num2<=10)))
print((num1 >= 50 or (num2<=10)))

# 부정 연산자
print(not(num1 >= 50))

```

```python
# 특정 글자의 모든 위치를 알아내는 법
str = "I have a dream."
seach_word = "a"
start = 0
count = 0

while True :
  position = str.find(seach_word, start)
  if position == -1:
    break
  count += 1
  print(f"단어 {seach_word}의 {count}번의 위치는 {position}입니다.")
  start = position + len(seach_word)
```

### 날짜형 변환

```python
from datetime import datetime, timedelta
```

```python
time1 = datetime(2020,12,22,21,0,0,0)
print(time1)
```

```python
type(time1)
```

```python
# 날짜 계산 하기
time2 = datetime.now()
print(time2)
print(time2 - time1)
```

```python
print((time2- time1).days,"일")
print((time2- time1).seconds,"초") # 일은 빼놓고 나머지 계산
```

```python
# 시간으로 확인하기
((time2 - time1).days * 24) + math.floor((time2 - time1).seconds / 60 / 60)
```

```python
# 특정 날짜로 부터 시간 계산
print("현재 시간부터 5일 뒤:", time2 + timedelta(days=5))
print("현재 시간부터 3일 전:", time2 + timedelta(days=-3))
print("현재 시간부터 1일 뒤의 2시간 전", time2 + timedelta(days=1,hours=-2))
```
