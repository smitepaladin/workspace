### Class

```python
class Calc():
  def setdata(self, first, second):
    # first = second = 0 # property, 안적어도 된다. 안적는게 퍼포먼스가 좋다.
    self.first = first
    self.second = second
  def sum(self):
    return self.first + self.second
  def sub(self):
    return self.first - self.second
  def mul(self):
    return self.first * self.second
  def div(self):
    return self.first / self.second
```

```python
calc01 = Calc()
calc01.setdata(4,2)
print(f"덧셈 : {calc01.sum()}")
print(f"뺄셈 : {calc01.sub()}")
print(f"곱셈 : {calc01.mul()}")
print(f"나눗셈 : {calc01.div()}")
```

```python
class Calc():
  # Property
  first = 0
  second = 0

  # Constructor
  def __init__(self, first, second):
    self.first = first
    self.second = second
  
  # Method
  def sum(self):
    return self.first + self.second
  def sub(self):
    return self.first - self.second
  def mul(self):
    return self.first * self.second
  def div(self):
    return self.first / self.second
```

```python
calc01 = Calc(4,3)
print(f"덧셈 : {calc01.sum()}")
print(f"뺄셈 : {calc01.sub()}")
print(f"곱셈 : {calc01.mul()}")
print(f"나눗셈 : {calc01.div()}")
```

### 상속

```python
class Add():
  num1 = 0
  num2 = 0

  def __init__(self, num1, num2):
    self.num1 = num1
    self.num2 = num2


  def add(self):
    return self.num1 + self.num2
```

```python
addition = Add(1,2)
print(addition.add())
```

```python
# 상속받기
class Sub(Add):
  num1 = 0
  num2 = 0

  def __init__(self, num1, num2):
    self.num1 = num1
    self.num2 = num2


  def sub(self):
    return self.num1 - self.num2
```

```python
subtraction = Sub(2,3)
print(subtraction.add())
```

```python
class Mul(Sub, Add):
  num1 = 0
  num2 = 0

  def __init__(self, num1, num2):
    self.num1 = num1
    self.num2 = num2


  def mul(self):
    return self.num1 * self.num2
```

```python
multiplation = Mul(2,3)
print(multiplation.mul())
```
