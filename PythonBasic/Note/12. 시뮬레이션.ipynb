### 시뮬레이션 (몬테카를로 시뮬레이션)
: 자료가 없을 경우 난수를 통해서 가설 증명

##### 동전의 앞면과 뒷면이 나올 확률이 같다는 것을 증명

```python
import random
```

```python
random.randint(1,2)
```

```python
# 난수를 통한 방법 알아보기
x = random.randint(1,2)
if x == 1:
  result = "앞면"
else:
  result = "뒷면"

print(result)
```

```python
# 동전의 앞면과 뒷면 결과값을 함수로 만들기

def coinF(n):
  result_front = result_back = 0
  for i in range(1,n+1):
    x = random.randint(1,2)
    if x == 1:
      result_front += 1
    else:
      result_back += 1
  print(f"{n}번 시도: 앞면 {result_front} / 뒷면 {result_back}")

```

```python
coinF(10)
coinF(100)
coinF(1000)
coinF(10000)
coinF(100000)
```

### 주사위 번호가 나올 확률이 같다는것을 증명

```python
def dice(n):
  result_1 = result_2 = result_3 = result_4 = result_5 = result_6 = 0
  for i in range(1, n+1):
    x = random.randint(1,6)
    match x:
      case 1:
        result_1 += 1
      case 2:
        result_2 += 1
      case 3:
        result_3 += 1
      case 4:
        result_4 += 1
      case 5:
        result_5 += 1
      case 6:
        result_6 += 1
  print(f"{n}번 시도시  ==>")
  print(f"1은 {result_1}번 검출 | 2은 {result_2}번 검출 | 3은 {result_3}번 검출")
  print(f"4은 {result_4}번 검출 | 5은 {result_5}번 검출 | 6은 {result_6}번 검출")
  print("-"*60)
      
```

```python
dice(10)
dice(100)
dice(1000)
dice(10000)
dice(100000)
```

```python
def dice(n):
  dice = [0,0,0,0,0,0]

  for i in range(1, n+1):
    x = random.randint(1,6)
    dice[x-1] += 1
  
  print(f"{n}번 시도시 -->")
  for i in range(0, 5+1, 3):
    print(f"{i+1} : {dice[i]} | {i+2} : {dice[i+1]} | {i+3} : {dice[i+2]}")
  print("-"*60)
```

```python
dice(10)
dice(100)
dice(1000)
dice(10000)
dice(100000)
```
