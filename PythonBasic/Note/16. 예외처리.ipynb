## 예외처리

```python
4 / 0
```

```python
try:
  4 / 1
except ZeroDivisionError as e:
  print(e)
```

```python
try:
  print(4 / 0)
except:
  print("Error")
```

```python
try:
  f = open("../Data/test.txt", "r")
except:
  print("Error")
else:
  data = f.read()
  f.close()
```

```python
try:
  f = open("../Data/new.txt", "r")
except:
  print("Error")
else:
  data = f.read()
  print(data)
finally:
  f.close()
```

---
### assert
: Exception을 주어진 Message로 변환

```python
age = int(input("나이 입력 :"))
assert age >= 0, "나이가 마이너스 인게 가능한가? 휴먼!"
print(f"당신의 나이는 {age} 살 입니다.")
```
