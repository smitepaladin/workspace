### file 읽고 쓰기
파일 객체 = open(파일 이름, 파일 열기 모드)   
파일 열기 모드 : r(읽기), w(쓰기 = del + inst), a(추가)

```python
# 파일 쓰기모드로 열어 출력값 저장하기.
f = open("../Data/new.txt", "w")
for i in range(1, 10+1):
  f.write(f"{i}번째 줄 입니다.\n")
f.close()
```

```python
# 파일을 읽기모드로 열어 파일 내용 출력하기 - Method 1

f = open("../Data/new.txt", "r")
while True:
  line = f.readline()
  if not line:
    break
  print(line, end="")
f.close
```

```python
# 파일을 읽기모드로 열어 파일 내용 출력하기 - Method 2

f = open("../Data/new.txt", "r")
lines = f.readlines()
f.close()
for line in lines:
  print(line, end="")
```

```python
# 파일을 읽기모드로 열어 파일 내용 출력하기 - Method 3

f = open("../Data/new.txt", "r")
data = f.read() # 메모리를 가져오자
f.close()
print(data)
```

```python
# 파일에 새로운 내용 추가하기
f = open("../Data/new.txt", "a")
for i in range(11, 20+1):
  f.write(f"{i}번째 줄 입니다.\n")
f.close()
```

```python
# 파일을 읽기모드로 열어 파일 내용 출력하기 - Method 3

f = open("../Data/new.txt", "r")
data = f.read() # 메모리를 가져오자
f.close()
print(data)
```

### with문
: 파일의 열고 닫음을 자동으로 처리하기 위해 with문을 벗어나면 자동으로 close()

```python
# 파일 만들기
with open("../Data/new2.txt", "w") as f:
  for i in range(1, 10+1):
    f.write(f"{i}번째 줄 입니다.\n")
```

```python
# 파일 읽기
with open("../Data/new2.txt", "r") as f:
  data = f.read() # 메모리를 가져오자
  print(data)
```
