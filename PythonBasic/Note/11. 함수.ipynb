### 사용자 정의 함수

```python
# 매개변수가 없는 함수
def hiF():
  return 'Hi'
```

```python
hiF()
```

```python
# 매개변수가 있는 함수
def testF(num1, num2):
  return num1 + num2, num1 - num2
```

```python
add, subtract = testF(2,3)
print(add, subtract)
```

```python
# 가변 길이 인수를 가진 함수 (unpacking 연산자)
def sumManyF(*nums):
  sum = 0
  for i in nums:
    sum += i
  return sum
```

```python
# 가변 길이 인수를 가진 함수 (unpacking  연산자)

print(sumManyF(1,2,3))
print(sumManyF(1,2,3,4,5,6,7,8))
```

```python
# 가변길이 키워드 매개변수
def print_items(**nums):
  for k, v in nums.items():
    print(k, v)
```

```python
print_items(a=1,b=2,c=3)
```

Ex : 구구단에서 단과 곱해지는 수를 리스트로 입력하여 출력하기

```python
def gugu(x,y):
  for num1 in x:
    for num2 in y:
      print(f"{num1} X {num2} = {num1 * num2}")
    print("-"*12)
```

### 덧셈 기능과 곱셈 기능을 구분하여 숫자의 결과를 보여주는 함수

```python
def sum_mulF(calc, *nums):
  if calc == 'sum':
    sum = 0
    for i in nums:
      sum += i
    print(sum)
  elif calc == 'mul':
    mul = 1
    for e in nums:
      mul *= e
    print(mul)
  else:
    error = "What are you talking about?"
    print(error)

```

```python
sum_mulF('sum', 1,2,3,4,5,6) # 21
sum_mulF('mul', 1,2,3,4,5) # 120
sum_mulF('abb', 1,2,3,4,5) # What are you talking about?
```

```python
# 특정한 결과값을 무시하고 싶은 경우
def say_nickF(nick):
  if nick == "바보":
    return
  return (f"나의 별명은 {nick} 입니다.")
```

```python
print(say_nickF("천재"))
```

```python
print(say_nickF("바보"))
```

### 함수에 초기값(defalt value) 설정하기

```python
def say_myself(name, old, gender = True): # 값이 안들어오면 기본값을 가져가겠다
  print(f"나의 이름은 {name}입니다.")
  print(f"나이는 {old}살 입니다.")
  print("남자 입니다." if gender else "여자 입니다.")
```

```python
say_myself("초선", 30, False)
```

```python
x = 0

def func1():
  x = 1 # x를 썼지만, 지역변수이다.
  print(x) 

def func2():
  print(x) # 전역변수 x값을 가져오게 된다.
```

```python
func1()
func2()
```

```python
x = 0

def func1():
  global x # 전역변수 선언
  x = 1
  print(x)

def func2():
  print(x)
```

```python
func1()
func2()
```

### 람다(lambda) 함수
- 보통 함수를 한줄로 간결하게 만들 때 사용
- return이 없다.

```python
def addFunction(a,b):
  return a+b
```

```python
addFunction(10,20)
```

```python
# 위의 함수를 람다함수로 변환
addLambda = lambda a, b : a + b # 입력값이 a,b가 들어오고 a + b를 하자.
addLambda(10,20)
```

### 중첩 함수(nested function) / 내부 함수(inner function)
: 함수 안에서 다른 함수를 정의

```python
def outer_function(x,y):
  def inner_function(z):
    return x + y + z
  return inner_function
```

```python
result = outer_function(1,2)
print(result(3))
print(result(4))
```

### 재귀(Recursion)
: 계산을 위해 자기 자신을 참조하는 방법

Factorial 함수

```python
# 반복 반식 / 가정은 n은 0보다 큰 정수
def fact_iter(n):
  result = 1
  for i in range(1, n+1):
    result *= i
  return result
```

```python
print(fact_iter(4))
```

```python
# 재귀 방식의 함수
def fact_rec(n):
  if n == 1:
    return n
  else:
    return n * fact_rec(n-1)
```

```python
print(fact_rec(4))
```
