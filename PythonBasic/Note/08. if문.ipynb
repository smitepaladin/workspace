```python
# 홀수 짝수 구분
x = 9

if x % 2 == 0:
  print(x, 'is even.')
else:
  print(x, 'is odd.')
```

```python
# 사용자로부터 숫자를 입력받아 홀짝수 구분
x = int(input("숫자를 입력 하세요"))

if x % 2 == 0:
  print(x, 'is even.')
else:
  print(x, 'is odd.')
```

## Ex
### 사용자로부터 점수를 입력받아 학점으로 출력하기
### 출력 예) 점수는 ___ 이고 학점은 __ 입니다.

```python

point = int(input("점수를 입력하세요"))

if point >= 90:
  grade = 'A'
elif point >= 80:
  grade = 'B'
elif point >= 70:
  grade = 'C'
elif point >= 60:
  grade = 'D'
else:
  grade = 'F'

print(f"점수는 {point}이고 학점은 {grade}입니다")
```

### match case

```python
x = 9

match x % 2:
  case 0:
    print(f"{x}는 짝수 입니다.")
  case 1:
    print(f"{x}는 홀수 입니다.")
```

```python
x = 9

match x:
  case x if x % 2 == 0:
    print("even")
  case _:
    print("Odd")
```

```python
var01 = int(input("input Number :"))

match var01:
  case 0:
    print("0")
  case 88 | 99:
    print("88 or 99")
  case var01 if var01 % 10 == 0:
    print("10의 배수")
  case var01 if var01 > 0:
    print("양수")
  case var01 if var01 < 0:
    print("음수")
  case _:
    print("~~~~")
```

### Ex
사용자로 부터 점수를 입력받아 학점으로 출력하기   
단) match case문으로 작성하기, 입력점수는 0~ 100 사이의 입력만 받는다

```python
score = int(input("input Score :"))

match score // 10:
  case 10:
    result = 'A'
  case 9:
    result = 'A'
  case 8:
    result = 'B'
  case 7:
    result = 'C'
  case 6:
    result = 'D'
  case _:
    result = 'F'

print(f"점수는 {score}이고 학점은 {result}입니다")
```

```python
# list의 내용을 if 문으로 확인하기
pocket = ['paper','smartphone']
card = 1

if 'money' in pocket:
  print('현금으로 택시 타기!')
elif card:
  print('신용카드를 이용한 택시 타기!')
else:
  print('튼튼한 다리 사용!')
```

### 삼항연산자
: 변수 = 참인경우 if(조건문) else 거짓인 경우

```python
num = 10
result = "짝수" if num % 2 == 0 else "홀수"
print(result)
```

```python
# 사용자로 부터 입력을 받아 입력 받은 수가 5보다 크면 입력수의 2배를 구하고 그렇지 않으면 2만 더하여 출력한다.

num1 = int(input("input number :"))

num1 = num1 * 2 if num1 > 5 else num1 + 2
print(num1)
```
