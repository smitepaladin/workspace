### Ex : 1부터 10000까지의 수중   
1) 3의 배수의 갯수 구하기
2) 5의 배수의 갯수 구하기
3) 3의배수 이면서 5의 배수인 갯수 구하기
4) 3의배수 이거나 5의 배수인 갯수 구하기

```python
data = range(1, 10000+1)

count3 = count5 = count3and5 = count3or5 = 0

for i in data:
  if i%3 == 0:
    count3 += 1
  if i%5 == 0:
    count5 += 1
  if (i%3 == 0) and (i%5 == 0):
    count3and5 += 1
  if (i%3 == 0) or (i%5 == 0):
    count3or5 += 1

print(count3, count5, count3and5, count3or5)
```

```python
# 검증
count3 + count5 - count3and5 == count3or5
```

### Ex


총 5명의 학생이 시험을 보았는데, 시험점수가 60점을 넘으면 합격이고
그렇지 않으면 불합격이다.    합격, 불합격 보여주기

```python
scores = [90,25,57,45,80]
```

```python
number = 1
for score in scores:
  print(f"{number}의 학생은 점수가 {score} {"이므로 합격" if score >= 60 else "이어서 불합격"} 입니다")
  number += 1
```

```python
# Ex : 60점 이상인 합격자 번호만 출력하기
scores = [90,25,57,45,80]
number = 0
passList = []

# 합격자 명단 구하기

for score in scores:
  number += 1
  if score < 60:
    continue
  else:
    passList.append(number)
    print(f"{number}의 학생은 점수가 {score} {"이므로 합격" if score >= 60 else "이어서 불합격"} 입니다")
print(passList)
```

### 문장과 단어 추출

```python
string = """나는 홍길동 입니다.
주소는 서울시 입니다.
나이는 35세 입니다."""
string 
```

```python
sents = []
words = []
```

```python
# 문단 -> 문장
for sen in string.split(sep="\n"):
  sents.append(sen)
  # 문장 -> 문단
  for word in sen.split():
    words.append(word)
```

```python
words
```

## 파이썬의 내포(Comprehension)함수

```python
var01 = [1,2,3,4]
var02 = []

for i in var01:
  var02.append(i * 5)

var02
```

```python
var02 = [i * 5 for i in var01]
var02
```

```python
var01 = [1,2,3,4]
var02 = [i * 5 for i in var01 if i % 2 == 0]
var02
```

```python
# 구구단 결과만 List로 만들기 (2단에서 ~ 9단)
result = [dan * i for dan in range(2,9+1) for i in range(1,9+1)]
print(result)
```

```python
import numpy as np
# 60점 이상의 학생 점수만 List로 만들어 평균 구하기
scores = [90,25,57,45,80]
passList = [score for score in scores if score >= 60]
np.mean(passList)
```

```python
# 구구단 결과만 List로 만들기 (2단에서 ~ 9단)
result = [[dan * i for i in range(1,9+1)] for dan in range(2,9+1)]
result
```

```python
import numpy as np
result = [np.sum([dan * i for i in range(1,9+1)]) for dan in range(2,9+1)]
print(result)
```
