### 리스트 관련 함수

```python
# 리스트에 데이터 추가
var = [1,2,3]
var.append(4)
var.append([5,6])
var
```

```python
# 리스트의 정렬
var = [1,4,3,2]

# 오름차순
var.sort()
print(var)

# 내림차순
var.reverse()
print(var)
```

```python
# 리스트에 데이터 삽입
var = [1,2,3]
var.insert(0,4)
var
```

```python
# 리스트에 요소 제거
var = [1,2,3,4,4,4,5]
var.remove(4)
print(var)
```

```python
# 리스트에서 데이터 끄집어 내기
var = [1,2,3]
print(var.pop()) # 리스트 마지막 데이터를 보여주고 해당 데이터는 삭제
print(var)
```

```python
# for문을 사용한 pop
var = [1,2,3]
for _ in range(len(var)):
  print(var.pop())
print(var)
```

```python
# 원하는 위치를 pop하기
var = [1,2,3]
print(var.pop(0))
print(var)
```

```python
# 리스트의 확장
var = [1,2,3,4]
var.extend([5,6])
var
```

### Ex

```python
# 1부터 10까지의 수 중 짝수 List와 홀수 List를 따로 생성하고
# 짝수 List에 홀수 List를 확장하여 합계 구하기

evenVar = list(range(2, 10+1, 2))
oddVar = list(range(1, 10+1, 2))
evenVar.extend(oddVar)

sum(evenVar)
```

### 리스트의 복사
1. Call by Reference

```python
var01 = [1,2,3,4]
var02 = var01
```

```python
var02[0] = 10
```

```python
print(var01)
print(var02)
```

### Call by Value

```python
var01 = [1,2,3,4]
var02 = var01.copy()
```

```python
var02[0] = 10
```

```python
print(var01)
print(var02)
```

```python
from copy import copy
```

```python
var01 = [1,2,3,4]
var02 = var01.copy()
var02[0] = 10
print(var01)
print(var02)
```

```python
var01 = [1,2,3,4]
var02 = var01[:]
var02[0] = 10
print(var01)
print(var02)
```
