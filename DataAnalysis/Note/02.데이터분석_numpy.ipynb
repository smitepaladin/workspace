```python
arr2 = np.array(
  [
    [7,8,9],
    [10,11,12]
  ]
)
arr2
```

###  배열 차원 변경

```python
arr = np.arange(1, 8+1, 1)
arr
```

```python
# 2차원 배열로 변경
arr1 = np.reshape(arr, (2,4))
arr1
```

```python
# arr을 3차원 배열로 변경
arr2 = np.reshape(arr, (2,2,2))
arr2
```
