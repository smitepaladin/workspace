### Series
: 데이터가 순차적으로 나열된 1차원 배열의 형태

```python
# Dictionary 만들기
dict_data = {
  'a' : 1,
  'b' : 2,
  'c' : 3,
}
dict_data
```

```python
import pandas as pd
```

```python
# Series로 변환
sr = pd.Series(dict_data)
sr
```

```python
sr['a']
```

### List => Series

```python
list_data = ['2015-12-31', 3.14, 'abc', 100, True]
list_data
```

```python
# Series로 변환
sr = pd.Series(
  list_data,
  index=['년도', '실수', '문자', '정수', '부울']
)
sr
```

```python
sr['년도']
sr.loc['년도']
sr.iloc[0]
```

### Tuple -> Series

```python
tuple_data = ('유비', '2025-12-31', '남', False)
tuple_data
```

```python
sr = pd.Series(
  tuple_data,
  index=['이름', '생년월일', '성별', '학생여부']
)
sr
```

```python
# sr['이름']
sr.loc[['생년월일', '성별']]
# sr.iloc[0]
```

```python
sr[['생년월일','학생여부']]
sr.loc[['생년월일', '학생여부']]
sr.iloc[[1,3]]
```

### Series의 연산

```python
student1 = pd.Series(
  {
    '국어' : 100,
    '영어' : 80,
    '수학' : 90,
  }
)
student1
```

```python
# 모든 과목의 점수를 100으로 나누자
student1 / 100
```

```python
student2 = pd.Series(
  {
    '수학' : 80,
    '국어' : 90,
    '영어' : 80,
  }
)
student2
```

```python
# 덧셈
addition = student1 + student2
addition
# index에 맞춰서 알아서 잘 계산한다.
```

```python
subtraction = student1 - student2
multiplication = student1 * student2
division = student1 / student2

subtraction
multiplication
division
```

```python
tbl = pd.DataFrame(
  [
    addition,
    subtraction,
    multiplication,
    division
  ],
  index=['덧셈', '뺼셈', '곱셈', '나눗셈']
)
tbl
```

### NaN(Not a Number:결측치)값이 있는 연산

```python
import numpy as np
```

```python
student1 = pd.Series(
  {
    '국어' : np.nan,
    '영어' : 80,
    '수학' : 90,
  }
)
student1
```

```python
student2 = pd.Series(
  {
    '수학' : 80,
    '국어' : 90,
  }
)
student2
```

```python
addition = student1 + student2
addition
```

```python
addition = student1.add(student2, fill_value=0)
subtraction = student1.sub(student2, fill_value=0)
multiplication = student1.mul(student2, fill_value=0)
division = student1.div(student2, fill_value=0)
```

```python
division
```
