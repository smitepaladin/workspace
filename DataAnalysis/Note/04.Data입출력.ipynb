```python
my_key = "AIzaSyCVgO5pFBFOmriF06-81_LmRXtSAwgrjxM"
```

```python
import googlemaps
```

```python
# 구글 맵스 객체 생성
maps = googlemaps.Client(key=my_key)
lat = []
lng = []
```

```python
# 장소(또는 주소) 리스트
places = ["서울시청", "국립국악원", "해운대해수욕장"]
```

```python
i = 0
lat.clear()
lng.clear()
for place in places:
    i += 1
    try:
        print(i, place)
        geo_loaction = maps.geocode(place)
        print(geo_loaction)
        lat.append(geo_loaction[0]['geometry']['location']['lat'])
        lng.append(geo_loaction[0]['geometry']['location']['lng'])
        
    except:
        lat.append('')
        lng.append('')
        print(i)
```

```python
# http://json.parser.online.fr/ 에서 확인
lat
```

```python
df = pd.DataFrame(
    {
      "위도" : lat,
      "경도" : lng
    },
    index=places
)
df
```
