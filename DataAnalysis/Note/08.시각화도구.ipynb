> 년도의 Data들이 object로 되어있다

```python
# 이동자수 (명) 날림
df.drop(
  0,
  axis='rows',
  inplace=True
)
```

```python
df.head()
```

```python
# Index 재정리
df = df.reset_index(drop=True)
df.head()
```

```python
#NaN값 정리 : NaN을 전(Prev)데이터로 채움
# df.fillna(method='ffill', inplace=true)
df.ffill(inplace=True)
df.head()
```

```python
# -으로 등록되어 있는 데이터의 갯수
len(df[df['1970'] == "-"])
```

```python
# - 가 아닌 데이터의 갯수
len(df[df['1970'] != "-"])
```

---
#### 년도별 컬럼에 대해 Object를 숫자로 변환

```python
import numpy as np
import warnings
warnings.filterwarnings('ignore')
```

```python
# - 을 NaN으로 변경
df = df.replace('-', np.nan)
df.head()
```

```python
# 년도별 컬럼의 값을 숫자로 변환하고 숫자로 변환이 안되는 값은 NaN으로 처리
df.iloc[:,2:] = df.iloc[:,2:].apply(pd.to_numeric, errors='coerce') # coerce가 숫자가 아니면 NaN으로 바꿔준다.

# NaN을 0으로 대체
df.fillna(0, inplace=True)
df.head()
```

```python
df.info()
```

```python
# 서울에서 다른 지역으로 이동한 데이터만 추출하여 정리
df_seoul = \
df[(df.전출지별 == '서울특별시') & (df.전입지별 != '서울특별시')]
df_seoul.head()
```

```python
# 전출지별 컬럼 삭제
df_seoul = df_seoul.drop(['전출지별'], axis='columns')
df_seoul.head()
```

```python
# 전입지별 -> 전입지로 컬럼 이름 변경
df_seoul.rename(
  {
    '전입지별' : '전입지'
  },
  axis='columns',
  inplace=True
)
df_seoul.head()
```

```python
# df_seoul의 index를 전입지로 변경
df_seoul.set_index(
  '전입지',
  inplace=True

)
df_seoul.head()
```

```python
# 서울에서 경기도로 이동한 인구 데이터 선택

sr_one = df_seoul.loc['경기도',:]
sr_one
```

```python
import matplotlib.pyplot as plt
```

```python
plt.plot(
  sr_one.index,
  sr_one.values
)
plt.show()
```

```python
plt.plot(sr_one)
plt.show()
# Series는 기본적으로 index가 x값, values가 y값이다. 입력하지 않아도.
```

### 부산광역시에서 서울로 전입한 그래프

### Style 추가해보기

```python
# Style
plt.style.use('ggplot')

# 그림 사이즈 지정(가로 14인치, 세로를 5인치)
plt.figure(figsize=(14, 5))

# x축 눈금 라벨 회전하기
plt.xticks(rotation = 45)


# Marker를 표시
plt.plot(sr_one, marker = 'o', markersize=10) # o, *, ^, +

# 차트 제목 추가

plt.title("서울 -> 경기 인구 이동")

# 축 이름 추가
plt.xlabel('기간', size=20)
plt.ylabel('이동 인구수', size=20)

# 범례
plt.legend(
  labels = ['서울 -> 경기'],
  loc='best',
  fontsize=15
)

# y축 범위 지정
plt.ylim(50000, 800000)

# 주석표시 - 화살표
plt.annotate(
  '',
  xy=(20, 620000), # 화살표의 머리부분
  xytext=(2, 290000), # 화살표의 꼬리부분
  xycoords='data', # 좌표 체계
  arrowprops=dict(arrowstyle='->', color='skyblue', lw = 5) # 화살표 서식
)

# 주석표시 - 텍스트
plt.annotate(
  '인구이동 증가(1997-1995)', # 텍스트입력
  xy=(10,400000), # 텍스트 위치 기준점
  rotation=23, # 텍스트 회전 각도
  va='baseline', # 텍스트 상하 정렬
  ha='center', # 텍스트 좌우 정렬
  fontsize = 12
)





# 주석표시 - 화살표
plt.annotate(
  '',
  xy=(47, 450000),
  xytext=(30, 580000),
  xycoords='data', # 좌표 체계
  arrowprops=dict(arrowstyle='->', color='Olive', lw = 5) # 화살표 서식
)

# 주석표시 - 텍스트
plt.annotate(
  '인구이동 감소(1995-2017)', # 텍스트입력
  xy=(38,520000), # 텍스트 위치 기준점
  rotation=-10, # 텍스트 회전 각도
  va='baseline', # 텍스트 상하 정렬
  ha='center', # 텍스트 좌우 정렬
  fontsize = 12
)




# 주석표시 - 화살표
plt.annotate(
  '',
  xy=(25, 600000),
  xytext=(28, 700000),
  xycoords='data', # 좌표 체계
  arrowprops=dict(arrowstyle='->', color='black', lw = 5) # 화살표 서식
)

# 주석표시 - 텍스트
plt.annotate(
  f'최대인구이동\n1995년\n{sr_one.values.max()}명', # 텍스트입력
  xy=(26,640000), # 텍스트 위치 기준점
  rotation=41, # 텍스트 회전 각도
  va='baseline', # 텍스트 상하 정렬
  ha='center', # 텍스트 좌우 정렬
  fontsize = 12
)

plt.show()
```
