#### 년도별 컬럼에 대해 Object를 숫자로 변환

```python
import numpy as np
import warnings
warnings.filterwarnings('ignore')
```

```python
# - 을 NaN으로 변경
df = df.replace('-', np.nan)
df.head()
```

```python
# 년도별 컬럼의 값을 숫자로 변환하고 숫자로 변환이 안되는 값은 NaN으로 처리
df.iloc[:,2:] = df.iloc[:,2:].apply(pd.to_numeric, errors='coerce') # coerce가 숫자가 아니면 NaN으로 바꿔준다.

# NaN을 0으로 대체
df.fillna(0, inplace=True)
df.head()
```

```python
df.info()
```

```python
# 서울에서 다른 지역으로 이동한 데이터만 추출하여 정리
df_seoul = \
df[(df.전출지별 == '서울특별시') & (df.전입지별 != '서울특별시')]
df_seoul.head()
```

```python
# 전출지별 컬럼 삭제
df_seoul = df_seoul.drop(['전출지별'], axis='columns')
df_seoul.head()
```

```python
# 전입지별 -> 전입지로 컬럼 이름 변경
df_seoul.rename(
  {
    '전입지별' : '전입지'
  },
  axis='columns',
  inplace=True
)
df_seoul.head()
```

```python
# df_seoul의 index를 전입지로 변경
df_seoul.set_index(
  '전입지',
  inplace=True

)
df_seoul.head()
```

```python
# 서울에서 경기도로 이동한 인구 데이터 선택

sr_one = df_seoul.loc['경기도',:]
sr_one
```

```python
import matplotlib.pyplot as plt
```

```python
plt.plot(
  sr_one.index,
  sr_one.values
)
plt.show()
```

```python
plt.plot(sr_one)
plt.show()
# Series는 기본적으로 index가 x값, values가 y값이다. 입력하지 않아도.
```

### 2개의 축을 갖는 그래프 작성
- 남북한 발전량 데이터셋을 사용
- 막대그래프는 북한의 발전량
- 선 그래프는 북한 발전량의 전년 대비 증감율을 백분율로 나타낸다.

```python
df = pd.read_excel("../Data/남북한발전전력량.xlsx")
df.head()
```

```python
# 북한 전력량만 선택
df = df.loc[5:8+1,:]
```

```python
df
```

```python
df.columns
# object다
```

```python
# df에서 전력량 (억kwh) 컬럼 삭제
# 하나 지울때는 del이 편함
del df['전력량 (억㎾h)']
df
```

```python
# index를 발전 전력별 컬럼으로 변경
df.set_index('발전 전력별', inplace=True)
df
```

```python
# 기준점을 발전 전력별로 변경
df = df.T
df.head()
```

```python
# 합계 - > 총발전량
df.rename({'합계':'총발전량'},axis='columns',inplace=True)
df.head()
```

```python
# 총발전량-1년 이라는 컬럼 생성
df['총발전량-1년'] = df.총발전량.shift(1)
df.head()
```

```python
# 증감율 컬럼 생성
df['증감율'] = ((df['총발전량'] / df['총발전량-1년']) - 1) * 100
df.head()
```

```python
# 2축 그래프 그리기

# pandas
ax1 = \
  df[['수력','화력']].plot(
    kind='bar',
    figsize=(20,10),
    width = 0.7,
    stacked=True
  )


ax2 = ax1.twinx() # x축은 같다. y축은 다를것이다.
# matplotlib
ax2.plot(
  df.index,
  df.증감율,
  ls='--', # 라인스타일
  marker='o',
  markersize=10,
  color='green',
  label='전년대비 증감율(%)'
)

ax1.set_ylim(0,500)
ax2.set_ylim(-50,50)



ax1.legend(loc='upper left')
ax1.set_title('북한 전력 발전량(1990 ~ 2016)', size=20)
ax1.set_xlabel('연도', size=15)
ax1.set_ylabel('발전량 (억㎾h)', size=12)
ax2.set_ylabel('전년 대비 증감율(%)', size=12)


x = df.index.get_loc('2015')
y = df.loc['2015','증감율'] + 5


# 주석표시 - 화살표
# plt.annotate(
#   '',
#   xy=(25, -12),
#   xytext=(25, 10),
#   xycoords='data', # 좌표 체계
#   arrowprops=dict(arrowstyle='->', color='black', lw = 5) # 화살표 서식
# )

# 주석표시 - 텍스트
plt.annotate(
  '2015년 대가뭄', # 텍스트입력
  xy=(x,y), # 텍스트 위치 기준점
  rotation=0, # 텍스트 회전 각도
  va='baseline', # 텍스트 상하 정렬
  ha='center', # 텍스트 좌우 정렬
  fontsize = 12
)


plt.show()
```
