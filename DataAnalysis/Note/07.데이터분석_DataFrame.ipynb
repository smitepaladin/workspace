### DataFrame Merging(병합)

```python
df1 = pd.DataFrame(
  {
    'key' : list('bbacaab'),
    'data' : range(7)
  }
)
df1
```

```python
df2 = pd.DataFrame(
  {
    'key' : list("abd"),
    'data' : range(3)
  }
)
df2
```

```python
# 병합하기
pd.merge(df1, df2, on='key')
# inner join
```

```python
# 모두 보이기
pd.merge(df1, df2, on='key', how='outer')
```

```python
# df1기준으로 합치기
pd.merge(df1, df2, on='key', how='left')
```

```python
df1 = pd.DataFrame(
  {
    'lkey' : list('bbacaab'),
    'data' : range(7)
  }
)
df1


df2 = pd.DataFrame(
  {
    'rkey' : list("abd"),
    'data' : range(3)
  }
)
df2
```

```python
# 서로 매치되는 키값이 없을 경우
pd.merge(df1, df2, left_on='lkey', right_on='rkey')
```

### Data Concatenating(연결)

```python
s1 = pd.Series(
  [0,1],
  index=["a","b"]
)
s1
```

```python
s2 = pd.Series(
  [2,3,4],
  index=["c","d","e"]
)
s2
```

```python
s3 = pd.Series(
  [5,6],
  index=["f","g"]
)
s3
```

```python
# s1, s2, s3 합치기

pd.concat(
  [s1,s2,s3]
)
```

```python
# Series를 합치면서 dataframe으로 만들기
pd.concat(
  [s1,s2,s3],
  axis='columns',
  sort=True
)
```

### DataFrame의 concatenate

```python
df1 = pd.DataFrame(
  np.arange(6).reshape(3,2),
  index=["a","b","c"],
  columns=['one','two']
)
df1
```

```python
df2 = pd.DataFrame(
  5 + np.arange(4).reshape(2,2),
  index=["a","c"],
  columns=['three','four']
)
df2
```

```python
# df1과 df2 합치기
pd.concat(
  [df1, df2],
  axis='columns'
)
```

### 데이터프레임 중복값 제거

```python
df = pd.DataFrame(
  {
    'names' : ['kenny','kenny','kenny','charles','charles'],
    'year'  : [2016, 2017, 2018, 2016, 2017],
    'points': [1.5, 1.77, 3.6, 2.4, 2.9]
  },
  columns=['year', 'names', 'points','penalty'], 
  index=['one', 'two', 'three', 'four', 'five']
)
df
```

```python
df.penalty = np.arange(0.1, 0.5+0.1,0.1)
df
```

```python
df['zeros'] = range(5)
df
```

```python
# Series
val = pd.Series(
  [-1.2, -1.5, -1.7,],
  index=['two','four','five']
)
val
```

```python
df['debt'] = val
df
```

```python
# net_points = points - penalty

df['net_points'] = df['points'] - df['penalty']
df
```

```python
# high_points : net_points가 2 초과면 True 아니면 False

df['high_point'] = df.net_points > 2
df
```

```python
# df의 high_points를 삭제
del df['high_point']
df
```

```python
del df['zeros']
del df['net_points']
df
```

```python
df.loc['six'] = [2018, 'Henry', 4, 0.1, 2.1]
df
```

```python
new_row = pd.DataFrame(
    [2018, 'Henry', 4, 0.1, 2.1],
    columns=['year', 'names', 'points', 'penalty', 'debt'],
    index=['six']
)

df = pd.concat([df, new_row])
df
```
