```python
# Fare(운임)이 20달러 이상인 데이터처음 5개

train[train['Fare'] >= 20].head()
```

```python
# Fare(운임)이 20달러 이상인 데이터는 몇개?

len(train[train['Fare'] >= 20])
```

##### 운임을 평균 이상으로 지불한 여성의 명수는?

```python
len(train[(train['Fare'] >= train['Fare'].mean()) & (train['Sex'] == 'female')])
```

##### 몇명의 여성이 C항구에서 승선하였는가?

```python
len(train[(train['Embarked'] == "C") & (train['Sex'] == 'female')])
```

```python
len(train[(train['Embarked'] == "C") & (train['Sex'] == 'male')])
```

```python
len(train[train['Embarked'] == "C"])
```

```python
len(train[(train['Embarked'] == "C") & (train['Sex'] == 'female')]) + len(train[(train['Embarked'] == "C") & (train['Sex'] == 'male')]) == len(train[train['Embarked'] == "C"])
```

##### 탑승항구가 C항구가 아닌 데이터의 갯수?

```python
len(train[(train['Embarked'] != "C")])
```

##### Q나 S인 항구에서 승선한 인원의 수는?

```python
len(train[(train['Embarked'] == "Q")]) + len(train[(train['Embarked'] == "S")])
```
