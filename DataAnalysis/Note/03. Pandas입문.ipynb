### 행 삭제

```python
df = pd.DataFrame(
  [
    [90,98,85,100],
    [80,89,95,90],
    [70,95,100,90],
  ],
  index=['서준','우현','인아'],
  columns=['수학','영어','음악','체육']
)
df
```

```python
df = pd.DataFrame(
  {
    '수학' : [90,80,70],
    '영어' : [98,89,95],
    '음악' : [85,95,100],
    '체육' : [100,90,90],
  },
  index=['서준','우현','인아']
)
df
```

```python
# DataFrame 복제
df2 = df.copy()
```

```python
# df2의 index가 우현인 1개행을 삭제
df2.drop(
  '우현',
  axis='rows', # 'columns'로 조절 가능
  inplace=True
)
df2
```

```python
df3 = df.copy()
```

```python
# df3의 index가 우현, 인아, 2개 행을 삭제
df3.drop(
  ['우현','인아'],
  axis='rows', # 'columns'로 조절 가능
  inplace=True
)
df3
```

### 행 선택
- loc : 행의 이름으로 선택
- iloc : 행의 정수형 index로 선택

```python
df.loc['서준',:]
```

```python
df.iloc[0,:]
```

```python
ddf = pd.DataFrame(
  {
    '이름' : ['서준', '우현', '인아'],
    '수학' : [90,80,70],
    '영어' : [98,89,95],
    '음악' : [85,95,100],
    '체육' : [100,90,90],
  },
)
ddf
```

### 범위 slicing
- loc.[시작:끝:증가치]
- iloc.[시작index:끝index+1:증가치]

```python
ddf.iloc[::2,:]
```

```python
ddf.iloc[0:2+1:2,:]
```

```python
# iloc로 내림차순

ddf.iloc[::-1,:]
ddf.iloc[2::-1,:]
```

```python
ddf.loc[::2,'수학':'체육':2]
```

### DataFrame 연산

```python
import seaborn as sns
```

```python
titanic = sns.load_dataset("titanic")
titanic.head()
```

```python
titanic.tail()
```

```python
titanic.shape
```

```python
titanic.info()
```

```python
# titanic에서 age, are의 2개열을 선택하여 데이터프레임 만들기
df = \
  pd.DataFrame(
    titanic,
    columns=['age', 'fare']
  )
df.head()
```

```python
df = titanic.loc[:,['age','fare']]
df.head()
```

```python
# df의 각 데이터를 10 증가한 데이터프레임 만들기

addition = df + 10
addition.head()
addition.info()
```

```python
# addition에서 df를 뺀 결과로 데이터 프레임 생성하기
subtraction = addition - df
subtraction.head()
subtraction.info()
```
