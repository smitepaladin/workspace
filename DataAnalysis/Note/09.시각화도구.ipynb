---
### 히스토그램(histogram)
: 데이터의 빈도수를 그래프로 표현

```python
plt.style.use('classic')
```

```python
df = pd.read_csv("../Data/auto-mpg.csv")
df.head()
```

```python
# mpg(연비) 열에 대한 히스토그램 그리기
df.mpg.plot(
  kind='hist',
  bins=20,
  color='coral',
  figsize=(10,5)
)

plt.title('Histogram')
plt.xlabel('mpg')


plt.show()
```

> 자동차의 연비값은 대부분 좌측으로 편향되어 있어서 연비가 낮은 구간에 집중되어 있는 경향을 보인다.

---
### 산점도
- 서로 다른 두 변수의 관계를 나타낸다.
- 이때 각 변수는 연속되는 값을 갖는다.

```python
plt.style.use('default')
```

```python
# 연비와 차중에 대한 산점도

df.plot(
  kind='scatter',
  x='weight',
  y='mpg',
  c='coral',
  s=10,
  figsize=(10,5)
)

plt.show()
```

> 자동차의 무게와 연비 사이의 관계를 산점도로 표현
자동차의 무게가 높을 수록 연비가 낮아지는 역상관을 표현하고 있다.

---
### 버블(bubble) 차트
: 산점도에 각 점의 크기에 특정 데이터로 크기를 결정한다.

```python
# cylinder 갯수의 정규화(상대적 비율 계산)를 한다.
cylinder_size = df.cylinders / df.cylinders.max() * 300
cylinder_size
```

```python
# 3개의 변수로 산점도 그리기
df.plot(
  kind='scatter',
  x='weight',
  y='mpg',
  c='coral',
  s=cylinder_size,
  figsize=(10,5),
  alpha=0.3
)

plt.show()
```

### 크기를 컬러로

```python
df.plot(
  kind='scatter',
  x='weight',
  y='mpg',
  c=cylinder_size,
  s=50,
  figsize=(10,5),
  cmap='viridis',
  marker="o"
)

plt.show()
```
